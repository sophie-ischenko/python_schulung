# Block 2: Benutzereingaben mit `input()` & Fehler-Sollbruchstelle 1

## Das Konzept von `input()`

Bisher waren alle Daten hart im Code codiert. Jetzt öffnen wir das Programm für den Anwender.

- **Syntax:** `variable = input("Aufforderungstext")`
- **Funktionsweise:** Python gibt den Text aus, stoppt die Programmausführung und wartet, bis der Benutzer Text eingibt und mit `Enter` bestätigt. Der eingegebene Text wird in der Variablen gespeichert.

````
benutzer = input("Bitte geben Sie Ihren Windows-Benutzernamen ein: ")
print(f"Eingabe registriert für Benutzer: {benutzer}")
````

## ⚡ Fehler-Sollbruchstelle 1: Der Additions-Absturz (TypeError)

Wir provozieren nun ganz bewusst einen Fehler, um zu verstehen, was `input()` im Hintergrund tut.

**Schritt 1:** Führe die folgende Zelle aus.
**Schritt 2:** Gib bei der Abfrage `25` ein und drücke `Enter`.
**Schritt 3:** Analysiere die rote Fehlermeldung.

Erwartete Fehlermeldung:
```
TypeError: can only concatenate str (not "int") to str
```

**Fehler-Diagnose:**
1. **Was bedeutet `str`?** String (Text).
2. **Was bedeutet `int`?** Integer (Ganzzahl).
3. **Was bedeutet `concatenate`?** Verknüpfen/Aneinanderhängen.
4. **Die Ursache:** `input()` liefert *immer* einen String zurück, selbst wenn der Benutzer `25` eingibt. Für Python lautet die Rechnung: `"25" + 10`. Python weigert sich, Text und Zahl zu addieren.

````
# ⚡ Fehler-Sollbruchstelle 1 — bewusst fehlerhaft, bitte ausführen und Fehlermeldung lesen
alter = input("Wie alt bist du? ")
alter_in_zehn_jahren = alter + 10
print(alter_in_zehn_jahren)
````

# Block 3: Typkonvertierung & Fehler-Sollbruchstellen 2 & 3

## Die Typkonvertierung (Casting)

Um den Fehler aus Block 2 zu beheben, müssen wir den String in eine Zahl umwandeln.

- `int(variable)`: Versucht, den Inhalt in eine Ganzzahl zu konvertieren.
- `float(variable)`: Versucht, den Inhalt in eine Nachkommazahl zu konvertieren.

**Die Reparatur (kompakte Schreibweise):** Wir schachteln die Befehle ineinander. Python arbeitet von innen nach außen.

````
alter = int(input("Wie alt bist du? "))
alter_in_zehn_jahren = alter + 10
print(f"In 10 Jahren bist du {alter_in_zehn_jahren} Jahre alt.")
````

## ⚡ Fehler-Sollbruchstelle 2: Die ungültige Zahl (ValueError)

Was passiert, wenn der Benutzer keine Zahlen eingibt, obwohl das Programm eine Ganzzahl erwartet?

**Schritt 1:** Führe die Zelle unten aus.
**Schritt 2:** Gib bei der Abfrage das Wort `dreißig` (in Buchstaben) oder `Admin` ein.
**Schritt 3:** Analysiere die Fehlermeldung.

Erwartete Fehlermeldung:
```
ValueError: invalid literal for int() with base 10: 'dreißig'
```

**Fehler-Diagnose:**
1. **ValueError:** Wert-Fehler.
2. **Die Ursache:** Der Befehl `int()` versucht, Buchstaben in eine mathematische Zahl umzuwandeln. Da das Wort `"dreißig"` keine Ziffern enthält, bricht Python ab. `int()` funktioniert nur, wenn der String wie eine Zahl aussieht (z. B. `"30"`).

````
# ⚡ Fehler-Sollbruchstelle 2 — bitte mit einem nicht-numerischen Wert ausführen (z. B. "dreißig")
alter = int(input("Wie alt bist du? "))
alter_in_zehn_jahren = alter + 10
print(f"In 10 Jahren bist du {alter_in_zehn_jahren} Jahre alt.")
````

## ⚡ Fehler-Sollbruchstelle 3: Die vergessene Klammer (SyntaxError)

Durch das Schachteln von Funktionen wie `int(input(...))` müssen wir zwei Klammern öffnen und auch wieder zwei schließen.

Erwartete Fehlermeldung:
```
SyntaxError: unexpected EOF while parsing
```
*(In moderneren Python-Versionen oft präziser: `SyntaxError: '(' was never closed`)*

**Fehler-Diagnose:**
1. **SyntaxError:** Grammatikfehler.
2. **unexpected EOF (End of File):** Python hat das Ende der Datei erreicht, während es noch nach der schließenden Klammer `)` gesucht hat.

*Hinweis: Da ein `SyntaxError` das Ausführen der Zelle komplett verhindert, ist der fehlerhafte Code unten als Kommentar hinterlegt. Zum Ausprobieren die `#` vor den beiden Zeilen entfernen und Zelle ausführen.*

````
# ⚡ Fehler-Sollbruchstelle 3 — Klammer entfernen zum Testen (führt zu SyntaxError)
# port = int(input("Geben Sie den Port ein (z. B. 80):"
# print(f"Port {port} wird geöffnet.")

# Korrigierte Version:
port = int(input("Geben Sie den Port ein (z. B. 80): "))
print(f"Port {port} wird geöffnet.")
````
