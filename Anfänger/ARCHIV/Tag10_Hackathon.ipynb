# Option 1: Das Retro-Tamagotchi ("PyPet")

**Konzept:** Ein virtuelles Haustier, das gepflegt werden muss. Es hat drei Statuswerte (Hunger, Energie, Laune) auf einer Skala von 0 bis 100.

**Spielmechanik:** In einer `while`-Schleife wählst du Aktionen (`Füttern`, `Spielen`, `Schlafen`). Jede Aktion verändert die Werte. Erreicht der Hunger 100 oder die Energie 0, ist das Spiel vorbei.

### 🎨 Schritt 1: ASCII-Art-Vorlagen laden

Führe diese Zelle zuerst aus – sie definiert die "Bilder", die je nach Zustand deines Haustiers angezeigt werden.


```python
PET_HAPPY = """
   /\_/\
  ( ^.^ )
   > ^ <
 [Miau! Mir geht es super!]
"""

PET_HUNGRY = """
   /\_/\
  ( o.o )
   > v <
 [Magen knurrt... Hunger!]
"""

PET_DEAD = """
   /\_/\
  ( x.x )
   > o <
 [Oh nein! Dein Haustier ist weggelaufen...]
"""
```

### 🛠️ Schritt 2: Das Code-Gerüst (Starter-Kit)

Dieses Gerüst ist schon lauffähig – aber die Aktionen `Spielen` (Option 2) und `Schlafen` (Option 3) fehlen noch! Nur `Füttern` (Option 1) ist bereits fertig eingebaut. Das ist deine erste Aufgabe: Ergänze an der markierten Stelle (`# HIER: Weitere Logik für Option 2 und 3 einbauen!`) die passenden `elif`-Zweige.

*Erinnerung aus der Projektbeschreibung:*
- **Füttern:** +20 Energie, −20 Hunger *(schon vorhanden)*
- **Spielen:** +20 Laune, −10 Energie, +10 Hunger
- **Schlafen:** +50 Energie, +10 Hunger

*Tipp: `global hunger, energie, laune` am Anfang von `main()` sorgt dafür, dass du innerhalb der Funktion die Variablen von außen verändern darfst – ohne das `global`-Schlüsselwort dürftest du sie in der Funktion nur lesen, nicht überschreiben.*


```python
# Werte initialisieren
hunger = 50
energie = 50
laune = 50

def zeige_status(hunger, energie, laune):
    print("\n--- STATUS ---")
    print(f"Hunger:  {hunger}/100")
    print(f"Energie: {energie}/100")
    print(f"Laune:   {laune}/100")

    # ASCII Art je nach Zustand ausgeben
    if hunger > 80:
        print(PET_HUNGRY)
    elif hunger >= 100 or energie <= 0:
        print(PET_DEAD)
    else:
        print(PET_HAPPY)

def main():
    global hunger, energie, laune
    print("Willkommen bei PyPet - Deinem digitalen Haustier!")

    while True:
        zeige_status(hunger, energie, laune)

        # Prüfen auf Game Over
        if hunger >= 100 or energie <= 0:
            print("Das Spiel ist aus.")
            break

        print("\nWas möchtest du tun?")
        print("[1] Füttern (+20 Energie, -20 Hunger)")
        print("[2] Spielen (+20 Laune, -10 Energie, +10 Hunger)")
        print("[3] Schlafen (+50 Energie, +10 Hunger)")
        print("[q] Beenden")

        wahl = input("Deine Wahl: ").lower()

        if wahl == "q":
            print("Tschüss!")
            break
        elif wahl == "1":
            hunger = hunger - 20
            energie = energie + 20
        # HIER: Weitere Logik für Option 2 und 3 einbauen!

        # Werte-Begrenzung (nicht unter 0, nicht über 100)
        hunger = max(0, min(hunger, 100))
        energie = max(0, min(energie, 100))
        laune = max(0, min(laune, 100))

if __name__ == "__main__":
    main()
```

# Option 2: Das Mini-DnD / Text-Adventure ("Python Dungeon")

**Konzept:** Ein rundenbasiertes Kampfspiel gegen einen Drachen im Verlies.

**Spielmechanik:** Spieler und Drache haben Lebenspunkte (HP). Du kannst `Angreifen` (Zufallsschaden simulieren), einen `Heiltrank trinken` oder `Fliehen`.

### 🎨 Schritt 1: ASCII-Art-Vorlagen laden


```python
SWORD = """
      /| ________________
O|===|* >________________>
      \|
"""

DRAGON = """
  <>=======()
  (/\___/\)   /\_/\
  |      |   ( o.o )
  ( ^_^* )    > ^ <
 [ROAARRR! Wer stört meine Ruhe?]
"""

TREASURE = """
  ._________________.
  | [T R E A S U R E]|
  |  ############### |
  |_________________|
 [Glückwunsch! Du hast gewonnen!]
"""
```

### 🛠️ Schritt 2: Das Code-Gerüst (Starter-Kit)

Dieses Gerüst ist bereits vollständig lauffähig – Angreifen, Heiltrank und Fliehen funktionieren schon, inklusive Gegenangriff des Drachens mit `random.randint()` für Zufallsschaden.

*Neu hier: das Modul `random`. Mit `random.randint(min, max)` erzeugst du eine zufällige Ganzzahl zwischen `min` und `max` (beide Grenzen eingeschlossen) – perfekt, um Kampfschaden nicht jedes Mal gleich hoch ausfallen zu lassen.*


```python
import random  # Erlaubt Zufallszahlen: random.randint(min, max)

player_hp = 100
dragon_hp = 80
potions = 3

def main():
    global player_hp, dragon_hp, potions
    print(DRAGON)
    print("Du betrittst die Höhle des Python-Drachen!")

    while player_hp > 0 and dragon_hp > 0:
        print(f"\nDeine HP: {player_hp} | Drachen-HP: {dragon_hp} | Tränke: {potions}")
        print("[1] Angreifen (Schwert)")
        print("[2] Heiltrank trinken (+30 HP)")
        print("[3] Fliehen")

        wahl = input("Deine Aktion: ")

        if wahl == "1":
            # Zufälligen Schaden berechnen
            schaden = random.randint(10, 25)
            dragon_hp = dragon_hp - schaden
            print(SWORD)
            print(f"Treffer! Du fügst dem Drachen {schaden} Schaden zu.")

        elif wahl == "2":
            if potions > 0:
                player_hp = player_hp + 30
                potions = potions - 1
                print("Du trinkst einen Heiltrank. +30 HP.")
            else:
                print("Keine Tränke mehr übrig!")
                continue

        elif wahl == "3":
            print("Du rennst schreiend davon... Spiel vorbei.")
            break

        # Gegenangriff des Drachen (wenn er noch lebt)
        if dragon_hp > 0:
            drachen_schaden = random.randint(12, 22)
            player_hp = player_hp - drachen_schaden
            print(f"Der Drache speit Feuer und fügt dir {drachen_schaden} Schaden zu!")

    # Spielende auswerten
    if dragon_hp <= 0:
        print(TREASURE)
        print("Der Drache ist besiegt! Du plünderst den Schatz.")
    elif player_hp <= 0:
        print("\nDu wurdest im Kampf besiegt... Versuche es nächstes Mal wieder!")

if __name__ == "__main__":
    main()
```

# Option 3: Der Cyber-Attacken-Simulator ("CyberGuard")

**Konzept:** Ein Terminal-Simulator, bei dem du als Support-Mitarbeiter Angriffe abwehren musst.

**Spielmechanik:** Die Firewall-Integrität sinkt kontinuierlich. Du musst Aktionen wählen (`Port patchen`, `IP blockieren`, `Log-Analyse`), um die Integrität wieder hochzupumpen, bevor sie 0 % erreicht – und bevor das Bedrohungslevel 100 % erreicht.

### 🎨 Schritt 1: ASCII-Art-Vorlagen laden


```python
SHIELD = """
     /\\
    /  \\
   /====\\
  /      \\
 [FIREWALL: AKTIV & SICHER]
"""

ATTACK = """
  💥 [ALARM: ANGRIFF ERKANNT] 💥
      !!!!!!!!!
      !! 404 !!
      !!!!!!!!!
"""

CRASH = """
  _____________________
 [   SYSTEM CRASHED    ]
 [      GAME OVER      ]
  ~~~~~~~~~~~~~~~~~~~~~
"""
```

### 🛠️ Schritt 2: Das Code-Gerüst (Starter-Kit)

Dieses Gerüst ist vollständig lauffähig. Beachte die Spiellogik genau: **Jede Runde** steigt die Bedrohung automatisch und die Firewall-Integrität sinkt automatisch (durch zwei `random.randint()`-Aufrufe ganz am Ende der Schleife) – unabhängig davon, welche Aktion du gewählt hast. Du kämpfst also gegen die Zeit.


```python
import random

integrity = 100
threat_level = 10
patched_ports = []

def main():
    global integrity, threat_level, patched_ports
    print(SHIELD)
    print("CyberGuard Terminal gestartet. Schütze die Server!")

    while integrity > 0 and threat_level < 100:
        print(f"\nFirewall-Integrität: {integrity}% | Bedrohungslevel: {threat_level}%")
        print(f"Gepatchte Ports: {patched_ports}")
        print("[1] Port patchen")
        print("[2] IP-Sperre einrichten")
        print("[3] Log-Analyse (Bedrohung senken)")

        wahl = input("Kommando: ")

        if wahl == "1":
            port = input("Welchen Port patchen? (z.B. 80): ")
            patched_ports.append(port)
            integrity = integrity + 15
            print(f"Port {port} erfolgreich geschlossen.")

        elif wahl == "2":
            integrity = integrity + 20
            threat_level = threat_level - 10
            print("Bösartige IPs blockiert. Bedrohung sinkt.")

        elif wahl == "3":
            bedrohungs_senkung = random.randint(15, 30)
            threat_level = threat_level - bedrohungs_senkung
            print(f"Logs analysiert. Bedrohung um {bedrohungs_senkung}% gesenkt.")

        # Jede Runde erhöht sich die Bedrohung und beschädigt die Firewall
        threat_increase = random.randint(10, 25)
        threat_level = threat_level + threat_increase
        integrity = integrity - random.randint(10, 20)

        print(ATTACK)
        print(f"Neuer Angriff verzeichnet! Bedrohung steigt um {threat_increase}%.")

        # Werte begrenzen
        threat_level = max(0, min(threat_level, 100))
        integrity = max(0, min(integrity, 100))

    if integrity <= 0:
        print(CRASH)
        print("Die Hacker haben die Server übernommen!")
    elif threat_level >= 100:
        print("Sicherheitsnetz überlastet!")
    else:
        print("Bedrohung abgewehrt!")

if __name__ == "__main__":
    main()
```
