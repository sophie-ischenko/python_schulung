# Block 1: if/else oder while? (09:00 – 09:45 Uhr)

## Theorie: Der konzeptionelle Unterschied

Bevor du neuen Stoff lernst, klärst du heute eine fundamentale Design-Frage beim Programmieren: **Wann brauchst du eine einfache Bedingung (`if`) und wann eine Schleife (`while`)?**

Das ist eine Entscheidung, die du als Programmierer/in bei jeder neuen Aufgabe triffst, bevor du überhaupt die erste Zeile Code schreibst. Es lohnt sich also, dieses Gefühl früh zu trainieren.

### `if/else` – die einmalige Prüfung

Der Computer prüft die Bedingung **genau einmal**. Ist sie wahr, wird der zugehörige Code ausgeführt. Ist sie falsch, wird stattdessen (falls vorhanden) der `else`-Block ausgeführt. Danach läuft das Programm einfach an der nächsten Zeile weiter – die Bedingung wird nicht noch einmal angeschaut.

> **Metapher:** Eine Ampel auf deinem Arbeitsweg. Du schaust einmal hin, reagierst (anhalten oder weiterfahren) und fährst dann weiter. Du stehst nicht die ganze Zeit an der Ampel und schaust im Sekundentakt hin.

### `while` – die wiederkehrende Prüfung

Der Computer prüft die Bedingung. Ist sie wahr, führt er den Code im Schleifenkörper aus und springt danach **wieder nach oben**, um die Bedingung erneut zu prüfen. Das wiederholt sich so lange, bis die Bedingung irgendwann unwahr wird. Erst dann geht es im Programm nach der Schleife weiter.

> **Metapher:** Das Warten an einer geschlossenen Bahnschranke. Du prüfst immer wieder ("Ist sie noch zu?"), und solange die Antwort "ja" lautet, wartest du weiter. Erst wenn sich die Bedingung ändert ("Sie ist offen!"), gehst du weiter.

### Die Faustregel

- Passiert etwas **genau einmal**, wenn eine Bedingung zutrifft? → `if/else`
- Muss etwas **wiederholt** geprüft oder wiederholt ausgeführt werden, bis sich ein Zustand ändert? → `while`

### Code-Vergleich: Server-Speicher

Schau dir den Unterschied direkt am Code an. Beide Beispiele beschäftigen sich mit demselben Thema (Speicherplatz auf einem Server), lösen aber unterschiedliche Probleme.


````
# --- VERSION A: Einmaliger Check (if) ---
# Wir prüfen EINMAL, ob der Speicher zu voll ist, und geben EINMAL eine Warnung aus.
belegter_speicher = 95  # Prozent

if belegter_speicher > 90:
    print("WARNUNG: Einmaliger Alarm gesendet!")
````

````
# --- VERSION B: Fortlaufendes Monitoring (while) ---
# Wir simulieren ein Löschskript, das den Speicher SCHRITTWEISE leert,
# und prüfen nach jedem Schritt erneut, ob wir schon fertig sind.
belegter_speicher = 95  # Prozent

while belegter_speicher > 90:
    print(f"Speicher ist bei {belegter_speicher}%. Bereinigung läuft...")
    belegter_speicher = belegter_speicher - 3  # Speicher sinkt

print(f"Bereinigung abgeschlossen. Speicher bei {belegter_speicher}%.")
````

### Warum das wichtig ist

Ein sehr häufiger Anfängerfehler ist es, ein `if` zu benutzen, wo eigentlich ein `while` gebraucht wird (z. B. bei einer Benutzereingabe, die so lange wiederholt werden soll, bis sie korrekt ist) – oder umgekehrt eine `while`-Schleife zu bauen, wo eine einmalige Prüfung völlig ausreichen würde. Wenn du dir vor dem Programmieren kurz die Frage *"Passiert das nur einmal oder wiederholt sich das?"* stellst, ersparst du dir viele Debugging-Sitzungen.

# Block 2: Einführung in Funktionen (09:45 – 10:30 Uhr)

## Theorie: Warum Funktionen?

Bisher hast du Code geschrieben, den du bei wiederholtem Bedarf kopieren musstest – zum Beispiel, wenn du eine Berechnung an drei verschiedenen Stellen im Programm gebraucht hast. Das ist mühsam und fehleranfällig: Wenn sich die Berechnung ändert, musst du sie an jeder einzelnen Kopierstelle anpassen.

Das verstößt gegen ein zentrales Prinzip der Programmierung: **DRY** – **D**on't **R**epeat **Y**ourself ("Wiederhole dich nicht").

Eine **Funktion** ist eine benannte "Code-Kiste". Du definierst sie **einmal**, und danach kannst du sie im gesamten Skript **beliebig oft** mit ihrem Namen aufrufen – ohne den Code jedes Mal neu zu schreiben.

### Der Vorteil auf einen Blick

- **Wiederverwendbarkeit:** Einmal schreiben, überall nutzen.
- **Wartbarkeit:** Änderungen musst du nur an einer Stelle vornehmen.
- **Übersicht:** Dein Hauptprogramm wird lesbarer, weil Details in Funktionen "versteckt" sind.

## Theorie: Syntax einer Funktion

Eine Funktionsdefinition besteht aus folgenden Bausteinen:

- **`def`**: Leitet die Definition ein (Kürzel für "define" – definieren).
- **Name der Funktion**: Wie bei Variablen – klein geschrieben, mit Unterstrichen bei mehreren Wörtern (z. B. `berechne_summe`).
- **`()` / Parameter**: In den Klammern stehen die Variablen, die die Funktion von außen entgegennehmen kann. Das sind quasi die "Zutaten", die du der Funktion mitgibst.
- **`:`**: Der Doppelpunkt leitet den Funktionskörper ein, genau wie bei `if` oder `while`.
- **Einrückung**: Der gesamte Code, der zur Funktion gehört, muss eingerückt sein (in Colab automatisch, wenn du nach dem Doppelpunkt Enter drückst).

```
def funktionsname(parameter1, parameter2):
    # hier steht der Code, der bei jedem Aufruf ausgeführt wird
    ...
```

Wichtig: Wenn du eine Funktion nur **definierst**, passiert noch nichts! Der Code in der Funktion wird erst ausgeführt, wenn du die Funktion **aufrufst** – also ihren Namen mit Klammern in deinem Code schreibst.

### Beispiel


````
# Definition der Funktion
def begruesse_user(username):
    print(f"Hallo {username}, Ihr System ist bereit.")

# Aufruf der Funktion mit verschiedenen Werten
begruesse_user("Alex")
begruesse_user("Sabine")
````

**Was passiert hier genau?**

1. Die `def`-Zeile definiert die Funktion `begruesse_user` mit einem Parameter `username`. An dieser Stelle wird noch nichts ausgegeben.
2. Beim Aufruf `begruesse_user("Alex")` wird der Wert `"Alex"` in den Parameter `username` "hineingelegt", und der Funktionskörper wird ausgeführt.
3. Beim zweiten Aufruf `begruesse_user("Sabine")` läuft derselbe Code noch einmal ab, diesmal mit `"Sabine"` als `username`.

So ersparst du dir, den `print`-Befehl mit dem Text zweimal auszuschreiben – die Logik steht nur einmal in der Funktion.

# Block 3: Funktionen mit Rückgabewerten – `return` (10:45 – 11:30 Uhr)

## Theorie: `print()` vs. `return`

Das ist der **wichtigste Unterschied für Einsteiger** im Umgang mit Funktionen, also nimm dir hier ruhig etwas Zeit.

- **`print()`** gibt einen Text auf dem Bildschirm aus. Das ist praktisch, um etwas *anzuzeigen* – aber das Programm selbst kann mit diesem ausgegebenen Text **nicht weiterarbeiten**. Er ist nur für deine Augen gedacht, nicht für den Code.
- **`return`** schickt ein Ergebnis aus der Funktion **zurück** in das Hauptprogramm. Erst dadurch kannst du das Ergebnis in einer Variablen speichern und später weiterverwenden – z. B. in einer weiteren Rechnung.

Stell es dir so vor: `print()` ist wie ein Mitarbeiter, der dir das Ergebnis laut vorliest, es danach aber wegwirft. `return` ist wie ein Mitarbeiter, der dir einen Zettel mit dem Ergebnis in die Hand gibt – den kannst du danach weiterverwenden, in einer Schublade aufbewahren oder mit einer anderen Zahl verrechnen.

### Beispiel im Vergleich


````
# --- VERSION A: Nur print (kein Weiterarbeiten möglich) ---
def addiere_print(a, b):
    print(a + b)

addiere_print(5, 5)  # Gibt 10 aus – aber wir können mit der 10 nichts mehr tun.
````

````
# --- VERSION B: Mit return (Ergebnis wird zurückgeliefert) ---
def addiere_return(a, b):
    return a + b

ergebnis = addiere_return(5, 5)  # Gibt NICHTS aus, speichert aber 10 in 'ergebnis'
print(ergebnis)       # Erst hier sehen wir das Ergebnis
print(ergebnis * 2)   # Wir können mit dem Wert WEITERRECHNEN! (Gibt 20 aus)
````

### Merksatz

> **`print` zeigt – `return` liefert.** Willst du mit dem Ergebnis einer Funktion später noch etwas anfangen (weiterrechnen, in einer Liste speichern, in einer Bedingung prüfen), brauchst du `return`. `print` reicht nur, wenn du das Ergebnis lediglich anschauen willst und danach nichts mehr damit vorhast.

Ein Tipp fürs spätere Debuggen: Wenn du eine Funktion aufrufst und das Ergebnis in einer Variablen speichern willst (`x = meine_funktion(...)`), die Variable danach aber `None` enthält, hast du sehr wahrscheinlich vergessen, in der Funktion ein `return` statt (oder zusätzlich zu) `print` zu verwenden.

### Kleine Kontrollfrage zum Ausprobieren

Was gibt der folgende Code wohl aus? Überlege zuerst und führe dann die Zelle aus, um zu prüfen, ob du richtig lagst.


````
def verdopple_print(zahl):
    print(zahl * 2)

def verdopple_return(zahl):
    return zahl * 2

a = verdopple_print(4)
b = verdopple_return(4)

print("a ist:", a)
print("b ist:", b)
````

# Block 5: Code-Review & Ausblick (12:45 – 13:00 Uhr)

## Gemeinsame Fehleranalyse

Zum Abschluss schaut ihr euch gemeinsam typische Stolperfallen beim Einstieg in Funktionen an:

- **Vergessene Einrückung:** Der Code innerhalb der Funktion muss eingerückt sein. Fehlt die Einrückung, gehört die Zeile nicht mehr zur Funktion – Python meldet dann meist einen `IndentationError` oder führt den Code an der falschen Stelle aus.
- **Variablen außerhalb der Funktion nutzen wollen (Scope):** Eine Variable, die *innerhalb* einer Funktion erstellt wird, existiert nur dort – man nennt das ihren **Scope** (Gültigkeitsbereich). Versuchst du, außerhalb der Funktion auf diese Variable zuzugreifen, bekommst du einen `NameError`. Das ist einer der häufigsten Anfängerfehler!

Schau dir das folgende Beispiel an und überlege, warum die letzte Zeile einen Fehler wirft:


````
def berechne_quadrat(zahl):
    ergebnis = zahl * zahl  # 'ergebnis' existiert nur INNERHALB der Funktion
    return ergebnis

x = berechne_quadrat(4)
print(x)  # Das funktioniert: x wurde außerhalb, beim Aufruf, mit dem return-Wert befüllt

# Die folgende Zeile würde einen Fehler werfen, wenn du sie einkommentierst:
# print(ergebnis)  # NameError: 'ergebnis' ist nur innerhalb der Funktion bekannt!
````
