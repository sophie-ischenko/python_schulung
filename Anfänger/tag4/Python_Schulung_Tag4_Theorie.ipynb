# Block 2: Einführung in Listen (Theorie)

## Was ist eine Liste?

Bisher konnte eine Variable immer nur **einen einzigen Wert** speichern. Wenn man `hostname = "SRV-01"` schreibt, dann passt in diese Variable genau ein Servername — und für einen zweiten Server bräuchte man eine zweite Variable (`hostname2 = "SRV-02"`), dann eine dritte, und so weiter. Das funktioniert bei 2 oder 3 Servern noch halbwegs, aber bei 50, 500 oder 5000 Servern wäre das mit einzelnen Variablen komplett unpraktikabel.

**Die Analogie: Der Aktenschrank**

Eine Liste ist wie ein Aktenschrank mit nummerierten Schubladen. Anstatt für jedes Dokument einen eigenen, frei herumstehenden Tisch zu haben (= eine eigene Variable), packt man alle zusammengehörigen Dokumente in *einen* Schrank. Jede Schublade hat eine Nummer, und man kann gezielt "Schublade 3" öffnen, ohne die anderen anfassen zu müssen. Der Schrank selbst hat nur *einen* Namen (z. B. `server_liste`), kann aber beliebig viele Werte aufnehmen — und man kann später jederzeit neue Schubladen hinzufügen oder alte entfernen.

**Warum ist das in der IT so wichtig?**

Listen begegnen einem ständig: eine Liste aller aktiven Benutzer, eine Liste aller offenen Tickets, eine Liste aller IP-Adressen in einem Netzwerk. Ohne Listen (und Schleifen, die morgen folgen) müsste man für jeden einzelnen Server, jeden einzelnen Benutzer eigenen Code schreiben. Mit Listen definiert man stattdessen eine Datenstruktur, die beliebig wächst oder schrumpft.

**Syntax:** Eckige Klammern `[]` markieren Anfang und Ende der Liste. Die einzelnen Werte ("Elemente") werden durch Kommas getrennt.

Eine Liste kann grundsätzlich beliebige Datentypen enthalten — auch gemischt (`["SRV-01", 42, 3.14, True]` ist technisch gültig), auch wenn man in der Praxis meist versucht, in einer Liste möglichst "gleichartige" Dinge zu sammeln, damit der Code übersichtlich bleibt.

````
server_liste = ["SRV-01", "SRV-02", "SRV-03"]
print(server_liste)
````

## Zugriff über den Index

Genau wie bei Strings hat jedes Element in einer Liste eine feste, durchnummerierte Position, einen **Index**. Die Nummerierung beginnt bei `0`, nicht bei `1`. Das erste Element hat den Index `0`, das zweite den Index `1`, und so weiter.

```text
 Liste:     "192.168.1.1"   "192.168.1.2"   "192.168.1.3"
 Index+:          0               1               2
 Index-:         -3              -2              -1
```

Negative Indizes funktionieren genau wie bei Strings: `ips[-1]` liefert immer das *letzte* Element, egal wie lang die Liste ist — praktisch, um z. B. den zuletzt hinzugefügten Log-Eintrag abzurufen, ohne vorher zählen zu müssen.

`len()` (kurz für "length") verrät, wie viele Elemente aktuell in der Liste enthalten sind — wichtig, weil sich die Länge einer Liste jederzeit ändern kann.

````
ips = ["192.168.1.1", "192.168.1.2", "192.168.1.3"]
print(ips[0])   # Erstes Element
print(ips[-1])  # Letztes Element
print(len(ips)) # Anzahl der Elemente
````

## Elemente in einer Liste ändern

Zentraler konzeptioneller Unterschied zu Strings: **Strings sind unveränderlich (immutable)** — ein einzelnes Zeichen kann nicht nachträglich ausgetauscht werden, man kann nur einen komplett neuen String erzeugen (wie bei `.replace()`). **Listen dagegen sind veränderlich (mutable)** — ein einzelnes Element kann direkt an seiner Position überschrieben werden, ohne die ganze Liste neu zu erstellen.

Im Aktenschrank-Bild: Man tauscht einfach den Inhalt von Schublade Nummer 2 aus, ohne den gesamten Schrank neu zu bauen — die Adresse "Schublade 2" bleibt gleich, nur der Inhalt ändert sich.

Diese Eigenschaft macht Listen enorm flexibel für alles, was sich im Laufe eines Programms verändert — z. B. ein IT-Inventar oder eine Warteschlange von Support-Tickets.

````
hardware = ["Maus", "Tastatur", "Monitor"]
hardware[1] = "Headset"
print(hardware)  # ['Maus', 'Headset', 'Monitor']
````

# Block 3: List-Methoden & dynamische Veränderung (Theorie)

Bisher hatten die Listen immer eine feste Größe. In der Praxis ist das selten der Fall: Neue Server kommen hinzu, alte werden abgebaut; neue Tickets treffen ein, bearbeitete werden geschlossen. Python stellt dafür eingebaute **Methoden** zur Verfügung, mit denen sich Listen dynamisch, also zur Laufzeit, verändern lassen.

### Elemente hinzufügen — die Liste wachsen lassen

**`.append(wert)`** ist die mit Abstand am häufigsten verwendete Listen-Methode. Sie hängt einen neuen Wert ganz **hinten** an die bestehende Liste an — so, als würde man im Aktenschrank einfach eine weitere Schublade unten anbauen. Die Reihenfolge der vorhandenen Elemente bleibt dabei unverändert.

**`.insert(index, wert)`** erlaubt es, ein neues Element an einer **bestimmten Position** einzuschieben, statt es nur ans Ende zu hängen. Alle Elemente ab dieser Position rutschen automatisch eine Position weiter nach hinten.

Vergleich mit einer Warteschlange am Bahnschalter: `.append()` ist wie ein neuer Kunde, der sich hinten anstellt. `.insert(0, ...)` ist wie ein Kunde, der ganz vorne reingelassen wird, weil sein Anliegen dringend ist — alle anderen rücken automatisch eine Position nach hinten.

### Elemente entfernen — die Liste schrumpfen lassen

**`.remove(wert)`** durchsucht die Liste nach dem angegebenen **Wert** (nicht nach einer Position!) und löscht das *erste* Vorkommen davon. Kommt der Wert mehrfach vor, wird nur die erste Fundstelle entfernt. Ist der Wert gar nicht enthalten, wirft Python einen `ValueError`.

**`.pop(index)`** funktioniert umgekehrt: Man gibt eine **Position** an, und Python entfernt das Element an dieser Stelle. Zusatznutzen: `.pop()` **gibt den entfernten Wert zurück**, sodass man ihn direkt weiterverarbeiten kann — nützlich z. B. bei einer Ticket-Warteschlange. Ohne Angabe eines Index wird standardmäßig das *letzte* Element entfernt.

Merkhilfe: **`.remove()`** = "ich weiß, *was* weg soll, aber nicht wo es steht" — **`.pop()`** = "ich weiß, *wo* es steht, und möchte den entfernten Wert weiterverwenden".

````
tools = ["Ping", "Nslookup"]
tools.append("Tracert")
tools.insert(1, "Nmap")
print("Nach Hinzufügen:", tools)

tools.remove("Ping")
entfernt = tools.pop(0)
print("Verbleibend:", tools)
print("Entfernt wurde:", entfernt)
````

## Ausblick: typischer Fehler

`IndexError: list index out of range` tritt auf, wenn man versucht, auf einen Index zuzugreifen, den es in der Liste gar nicht gibt, etwa `liste[10]` bei einer Liste mit nur 5 Elementen.
