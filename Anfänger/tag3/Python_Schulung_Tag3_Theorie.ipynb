# Block 1.5: Der Code-Sanitäter (Debugging-Spezial)

> **Wichtigste Regel beim Debuggen:** Fehlermeldungen immer von unten nach oben lesen. Die letzte Zeile verrät meistens, was schiefgelaufen ist.

Nachfolgend vier reale Support-Bugs: jeweils der fehlerhafte Code mit erwarteter Fehlermeldung und Diagnose, danach die reparierte Version.

## Debug-Aufgabe D1: Der abgestürzte Ticket-Zähler (TypeError)

Erwartete Fehlermeldung:
```
TypeError: unsupported operand type(s) for -: 'str' and 'int'
```

**Diagnose:**
1. `offene_tickets` muss mit `int()` konvertiert werden.
2. Das Ergebnis `verbleibend` ist eine Zahl und kann nicht mit `+` an einen Text gehängt werden. Lösung: f-String.

````
# Fehlerhafter Code — bitte zuerst ausführen und Fehler lesen
offene_tickets = input("Wie viele offene Tickets gibt es? ")
geloeste_tickets = 5

# Hier soll die Differenz berechnet werden
verbleibend = offene_tickets - geloeste_tickets
print("Es verbleiben: " + verbleibend)
````

````
# Reparierter Code
offene_tickets = int(input("Wie viele offene Tickets gibt es? "))
geloeste_tickets = 5
verbleibend = offene_tickets - geloeste_tickets
print(f"Es verbleiben: {verbleibend}")
````

## Debug-Aufgabe D2: Der unvollständige Server-Check (SyntaxError & NameError)

Erwartete Fehlermeldung (Schritt 1):
```
SyntaxError: '(' was never closed
```
*(oder `unexpected EOF while parsing`)*

**Diagnose:**
1. Am Ende der `print`-Zeile fehlt die schließende Klammer `)` (SyntaxError).
2. Schließt man die Klammer, taucht der nächste Fehler auf: `NameError: name 'srv_Name' is not defined`. Python unterscheidet streng zwischen Groß- und Kleinschreibung (`srv_name` vs. `srv_Name`).

*Ein `SyntaxError` verhindert das Ausführen der gesamten Zelle. Der fehlerhafte Code ist daher unten als Kommentar hinterlegt — zum Ausprobieren die Kommentarzeichen entfernen und ausführen.*

````
# Schritt 1: fehlerhafter Code (auskommentiert, da SyntaxError die Zelle blockiert)
# srv_name = "SRV-Backup"
# status_aktiv = True
#
# print(f"Server {srv_Name} hat den Status: {status_aktiv}"

# Zum Testen: die drei Zeilen oben einkommentieren und Zelle ausführen -> SyntaxError: '(' was never closed
````

````
# Schritt 2: Klammer geschlossen, aber Tippfehler bleibt (führt zu NameError)
srv_name = "SRV-Backup"
status_aktiv = True

print(f"Server {srv_Name} hat den Status: {status_aktiv}")
````

````
# Schritt 3: vollständig reparierter Code
srv_name = "SRV-Backup"
status_aktiv = True
print(f"Server {srv_name} hat den Status: {status_aktiv}")
````

## Debug-Aufgabe D3: Der verunglückte Speicher-Rechner (ValueError)

Erwartete Fehlermeldung bei Eingabe von `2.5`:
```
ValueError: invalid literal for int() with base 10: '2.5'
```

**Diagnose:** Der Benutzer gibt eine Zahl mit Nachkommastelle ein. `int()` kann jedoch nur ganze Zahlen ohne Punkt verarbeiten. Die Konvertierung muss mit `float()` durchgeführt werden.

````
# Fehlerhafter Code — bitte mit Eingabe "2.5" ausführen
freier_speicher = int(input("Wie viel GB freier Speicher? "))
bedarf = 1.2
rest = freier_speicher - bedarf
print(f"Restlicher Speicher: {rest} GB")
````

````
# Reparierter Code
freier_speicher = float(input("Wie viel GB freier Speicher? "))
bedarf = 1.2
rest = freier_speicher - bedarf
print(f"Restlicher Speicher: {rest} GB")
````

## Debug-Aufgabe D4: Der "Division durch Null"-Fehler (ZeroDivisionError)

Erwartete Fehlermeldung:
```
ZeroDivisionError: division by zero
```

**Diagnose:** Mathematisch ist eine Division durch Null nicht definiert. Im IT-Alltag stürzen so Programme ab, wenn z. B. temporär keine Server aktiv sind. Behoben wird der Fehler, indem `aktive_server` auf einen Wert größer als Null gesetzt wird (z. B. `2`).

````
# Fehlerhafter Code
gesamtspeicher_gb = 500
aktive_server = 0

# Lastverteilung berechnen
last_pro_server = gesamtspeicher_gb / aktive_server
print(f"Jeder Server erhält {last_pro_server} GB.")
````

````
# Reparierter Code
gesamtspeicher_gb = 500
aktive_server = 2  # Muss ungleich 0 sein
last_pro_server = gesamtspeicher_gb / aktive_server
print(f"Jeder Server erhält {last_pro_server} GB.")
````

# Block 2: String-Indizierung & Slicing (Theorie)

Strings sind Ketten aus einzelnen Zeichen. Wir lernen, wie man gezielt Zeichen herausgreift.

```text
 Zeichen:   P   y   t   h   o   n
 Index+:    0   1   2   3   4   5
 Index-:   -6  -5  -4  -3  -2  -1
```

- **Indizierung:** Einzelne Zeichen über eckige Klammern ansprechen: `wort[0]` liefert `P`. Negative Indizes zählen von hinten: `wort[-1]` liefert das letzte Zeichen.
- **Slicing:** Bereiche ausschneiden mit `[start:stop]`. Der `stop`-Wert ist exklusive.
- **Schrittweite:** `[start:stop:step]` erlaubt es, Zeichen zu überspringen (z. B. `[::2]` für jedes zweite Zeichen).

````
wort = "Python"

print(wort[0])     # erstes Zeichen
print(wort[-1])    # letztes Zeichen
print(wort[0:3])   # die ersten drei Zeichen (Index 0,1,2)
print(wort[::2])   # jedes zweite Zeichen
print(wort[::-1])  # der ganze String rückwärts
````
