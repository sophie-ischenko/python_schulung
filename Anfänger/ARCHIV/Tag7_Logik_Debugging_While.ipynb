### Debug-Aufgabe D1: Die verschobene Einrückung (IndentationError)

**Diagnose:** Python definiert Code-Blöcke strikt über Einrückungen. Die Befehle nach dem `if:` müssen genau um dieselbe Anzahl an Leerzeichen (üblicherweise 4) eingerückt sein und auf derselben Ebene liegen. Hier ist die erste `print`-Zeile gar nicht eingerückt (obwohl sie zum `if`-Block gehören soll) und die zweite ist inkonsistent eingerückt — beides erzeugt einen `IndentationError`.

*Hinweis: Der fehlerhafte Code ist auskommentiert, da ein `IndentationError` sonst die ganze Zelle blockiert. Zum Ausprobieren die Kommentarzeichen entfernen.*

````python
# Debug-Aufgabe D1 — fehlerhafter Code (auskommentiert)
# speicher_voll = True
#
# if speicher_voll:
# print("Warnung: Speicher voll!")
#   print("Bitte Backup starten.")

# Reparierter Code:
speicher_voll = True
if speicher_voll:
    print("Warnung: Speicher voll!")
    print("Bitte Backup starten.")
````

### Debug-Aufgabe D2: Die vergessenen Doppelpunkte (SyntaxError)

**Diagnose:** In Python muss nach einer `if`- oder `else`-Bedingung zwingend ein Doppelpunkt `:` stehen, um den nachfolgenden Block einzuleiten. Hier fehlt der Doppelpunkt sowohl nach der `if`- als auch nach der `else`-Zeile.

*Hinweis: Auch hier ist der fehlerhafte Code auskommentiert, da ein `SyntaxError` die ganze Zelle blockiert.*

````python
# Debug-Aufgabe D2 — fehlerhafter Code (auskommentiert)
# status = "offline"
#
# if status == "offline"
#     print("Server ist offline.")
# else
#     print("Server läuft.")

# Reparierter Code:
status = "offline"
if status == "offline":
    print("Server ist offline.")
else:
    print("Server läuft.")
````

### Debug-Aufgabe D3: Die Zuweisung in der Bedingung (SyntaxError)

**Diagnose:** Ein einzelnes Gleichheitszeichen `=` ist eine Zuweisung. Für Vergleiche in Bedingungen muss zwingend das doppelte Gleichheitszeichen `==` verwendet werden. `if port = 80:` ist deshalb kein gültiger Vergleich, sondern ein `SyntaxError` — Python verhindert genau diese Verwechslung, indem es eine Zuweisung an dieser Stelle grundsätzlich nicht zulässt.

*Hinweis: Fehlerhafter Code wieder auskommentiert.*

````python
# Debug-Aufgabe D3 — fehlerhafter Code (auskommentiert)
# port = 80
#
# if port = 80:
#     print("Webserver aktiv.")

# Reparierter Code:
port = 80
if port == 80:
    print("Webserver aktiv.")
````

### Debug-Aufgabe D4: Logische Verwechslung (Logischer Fehler)

*Ziel:* Der Zugriff ist erlaubt, wenn der Benutzer `"Admin"` ODER `"Superuser"` heißt.

**Diagnose:** Dieser Code wirft keinen Fehler und lässt sich ganz normal ausführen — das macht ihn besonders tückisch, denn das Programm läuft fehlerfrei durch, liefert aber das *falsche* Ergebnis. Mit `and` verlangt Python, dass der Benutzername *gleichzeitig* `"Admin"` **und** `"Superuser"` ist. Das ist unmöglich, da eine Variable nicht gleichzeitig zwei verschiedene Werte haben kann — die Bedingung ist also immer `False`. Hier muss der logische Operator `or` verwendet werden.

````python
# Debug-Aufgabe D4 — fehlerhafter Code (läuft, liefert aber falsches Ergebnis)
username = "Superuser"

# Hier liegt der logische Fehler:
if username == "Admin" and username == "Superuser":
    print("Zugriff erlaubt.")
else:
    print("Zugriff verweigert.")
````

````python
# Debug-Aufgabe D4 — reparierter Code
username = "Superuser"
if username == "Admin" or username == "Superuser":
    print("Zugriff erlaubt.")
else:
    print("Zugriff verweigert.")
````

### Debug-Aufgabe D5: Der unberechenbare String (TypeError)

**Diagnose:** `input()` liefert *immer* einen String zurück, selbst wenn der Benutzer eine Zahl eingibt. Der Vergleich `ping < 50` versucht, Text mit einer Zahl zu vergleichen — Python weiß nicht, wie es einen String und eine Ganzzahl der Größe nach vergleichen soll, und stürzt mit einem `TypeError` ab. Die Eingabe muss zuerst in `int()` konvertiert werden.

````python
# Debug-Aufgabe D5 — fehlerhafter Code, bitte mit Eingabe "45" ausführen
ping = input("Geben Sie den Ping in ms ein: ")  # User gibt 45 ein

if ping < 50:
    print("Verbindung ist schnell.")
````

````python
# Debug-Aufgabe D5 — reparierter Code
ping = int(input("Geben Sie den Ping in ms ein: "))
if ping < 50:
    print("Verbindung ist schnell.")
````

### Debug-Aufgabe D6: Doppelte Verneinung (Logischer Fehler)

*Ziel:* Der Server soll patchen, wenn er *nicht* gesperrt ist (`gesperrt = False`).

**Diagnose:** Auch dieser Code läuft ohne Absturz durch, liefert aber (je nach Denkweise) ein verwirrendes oder falsches Ergebnis, weil er den Wahrheitswert doppelt verneint: Wenn `gesperrt` `False` ist, dann ist der Vergleich `gesperrt == False` bereits `True`. Das zusätzliche `not` davor dreht dieses `True` wieder zu `False` um — genau das Gegenteil von dem, was eigentlich gewollt war. Doppelte Verneinungen wie diese sind ein klassischer Stolperstein und lassen sich fast immer vereinfachen: `not gesperrt` reicht hier vollkommen aus.

````python
# Debug-Aufgabe D6 — fehlerhafter Code (läuft, aber mit falscher/verwirrender Logik)
gesperrt = False

if not gesperrt == False:
    print("Patch-Vorgang startet.")
else:
    print("Server ist gesperrt!")
````

````python
# Debug-Aufgabe D6 — reparierter Code
gesperrt = False
if not gesperrt:
    print("Patch-Vorgang startet.")
else:
    print("Server ist gesperrt!")
````

## 09:45 – 10:30 Uhr | Block 2: Einführung in die `while`-Schleife (Theorie)

### 1. Das Konzept von Schleifen — warum wir sie brauchen

Bisher mussten wir Code von Hand kopieren, wenn wir etwas mehrmals ausführen wollten. Wollten wir zum Beispiel viermal hintereinander "Ping gesendet" ausgeben, hätten wir die `print`-Zeile viermal abtippen müssen. Das funktioniert bei 4 Wiederholungen noch, aber was, wenn es 400 oder 40.000 sein sollen — oder wenn die genaue Anzahl vorher gar nicht bekannt ist (z. B. "frage so lange nach dem Passwort, bis es stimmt")? Genau für solche Fälle gibt es **Schleifen (Loops)**.

**Die Analogie: Der Wachmann an der Schranke**
Stellen Sie sich einen Wachmann an einer Parkplatzschranke vor. Er hat eine einzige, klare Regel: "Solange noch Autos in der Schlange warten, lasse das nächste durch." Er prüft diese Bedingung nicht nur einmal, sondern **immer wieder, vor jedem einzelnen Auto neu** — und erst wenn die Bedingung nicht mehr zutrifft (keine Autos mehr da), hört er auf. Genau so funktioniert eine `while`-Schleife: Sie prüft vor *jedem einzelnen* Durchlauf erneut, ob die Bedingung noch wahr ist, und wiederholt den Codeblock so lange, wie das der Fall ist.

### 2. Syntax der `while`-Schleife

```python
zaehler = 1

while zaehler <= 3:
    print(f"Durchlauf Nummer: {zaehler}")
    zaehler = zaehler + 1  # Wichtig: Zähler erhöhen!
```

Lassen Sie uns das im Detail nachvollziehen, Durchlauf für Durchlauf:
- **1. Prüfung:** Ist `zaehler <= 3`? `zaehler` ist `1`, also `1 <= 3` → `True`. Der Codeblock wird ausgeführt: Es wird `"Durchlauf Nummer: 1"` ausgegeben, und `zaehler` wird auf `2` erhöht.
- **2. Prüfung:** Ist `zaehler <= 3`? `zaehler` ist jetzt `2`, also `2 <= 3` → `True`. Ausgabe: `"Durchlauf Nummer: 2"`, `zaehler` wird auf `3` erhöht.
- **3. Prüfung:** Ist `zaehler <= 3`? `zaehler` ist `3`, also `3 <= 3` → `True`. Ausgabe: `"Durchlauf Nummer: 3"`, `zaehler` wird auf `4` erhöht.
- **4. Prüfung:** Ist `zaehler <= 3`? `zaehler` ist jetzt `4`, also `4 <= 3` → `False`. Die Schleife wird beendet, und das Programm läuft nach der Schleife weiter.

**Der entscheidende Baustein: Die Veränderung der Bedingung im Schleifenkörper.** Beachten Sie die Zeile `zaehler = zaehler + 1`. Ohne diese Zeile würde sich `zaehler` niemals verändern, die Bedingung `zaehler <= 3` wäre also *für immer* `True`, und die Schleife würde nie enden. Diese Zeile ist also nicht optional oder "nice to have" — sie ist der Mechanismus, der überhaupt dafür sorgt, dass die Schleife irgendwann aufhört.

### 3. ⚡ Achtung: Die Endlosschleife (Unendlicher Loop)

Wenn die Bedingung der Schleife *niemals* `False` wird — sei es, weil man vergisst, den Zähler zu erhöhen, oder weil man absichtlich `while True:` schreibt — läuft das Programm unendlich weiter, bis man es von außen stoppt. In einer echten Anwendung kann das den Arbeitsspeicher füllen oder das gesamte Programm einfrieren lassen.

```python
# Trainer zeigt diesen unendlichen Loop in Colab (NICHT ausführen, nur zur Anschauung):
# while True:
#     print("Ich laufe für immer!")
```

*Wichtiger Praxis-Hinweis:* In Google Colab (oder Jupyter) klickt man in so einem Fall sofort auf den **Stop-Button ("Interrupt execution")** links neben der Zelle, bzw. im Menü auf "Laufzeit" → "Ausführung unterbrechen". Man sollte niemals in Panik geraten — eine Endlosschleife lässt sich jederzeit von außen abbrechen, sie ist kein dauerhafter Schaden am System, sondern lediglich eine laufende Berechnung, die man stoppen muss.

Interessanterweise ist `while True:` selbst kein Fehler, sondern ein bewusst eingesetztes, sehr gebräuchliches Muster — nämlich immer dann, wenn wir die Schleife absichtlich von *innen* heraus beenden wollen, statt über eine Bedingung am Anfang. Genau das lernen wir jetzt in Block 3 mit `break`.

````python
# Live-Coding: Grundlegende while-Schleife
zaehler = 1

while zaehler <= 3:
    print(f"Durchlauf Nummer: {zaehler}")
    zaehler = zaehler + 1  # Wichtig: Zähler erhöhen!
````

## 10:45 – 11:30 Uhr | Block 3: Loop-Steuerung mit `break` und `continue` (Theorie)

Manchmal reicht die reine Bedingung am Kopf der `while`-Schleife nicht aus, um das gewünschte Verhalten abzubilden. Python stellt uns dafür zwei spezielle Sprungbefehle zur Verfügung, die wir *innerhalb* des Schleifenkörpers verwenden können, um den normalen Ablauf gezielt zu unterbrechen.

### 1. Schleife abbrechen mit `break`

Mit `break` beenden wir die Schleife **sofort und vollständig** — egal, ob die eigentliche Schleifen-Bedingung noch wahr wäre oder nicht. Man kann sich das wie einen Notausschalter vorstellen: Sobald `break` ausgeführt wird, springt das Programm direkt aus der Schleife heraus und macht mit dem Code danach weiter, ohne die Bedingung noch einmal zu prüfen.

Das ist besonders nützlich in Kombination mit `while True:` — einer Schleife, deren Kopf-Bedingung *immer* wahr ist und die sich daher niemals von selbst beendet. Solche Schleifen wirken auf den ersten Blick wie Endlosschleifen, sind es in der Praxis aber nicht, weil irgendwo *im Inneren* der Schleife ein `break` versteckt ist, das unter der richtigen Bedingung ausgelöst wird. Dieses Muster ist extrem verbreitet, etwa bei Menüs, Chatbots oder Login-Systemen, bei denen man vorher gar nicht weiß, wie oft die Schleife durchlaufen werden muss — man weiß nur, *unter welcher Bedingung* sie enden soll.

**Gemeinsames Live-Coding (Das unendliche Login):**
```python
# Wir fragen so lange nach dem Passwort, bis es korrekt ist
while True:
    eingabe = input("Passwort eingeben (oder 'exit' zum Abbrechen): ")
    if eingabe == "admin123" or eingabe == "exit":
        print("Schleife wird beendet.")
        break  # Springt sofort aus der Schleife
    print("Falsches Passwort! Versuchen Sie es erneut.")
```
Der Ablauf: `while True` bedeutet "wiederhole für immer". Erst die `if`-Bedingung *innerhalb* der Schleife entscheidet, wann tatsächlich Schluss ist — und genau dann sorgt `break` dafür, dass die "Endlosschleife" doch ein Ende findet.

````python
# Live-Coding: break — das unendliche Login
while True:
    eingabe = input("Passwort eingeben (oder 'exit' zum Abbrechen): ")
    if eingabe == "admin123" or eingabe == "exit":
        print("Schleife wird beendet.")
        break  # Springt sofort aus der Schleife
    print("Falsches Passwort! Versuchen Sie es erneut.")
````

### 2. Schritte überspringen mit `continue`

Während `break` die *gesamte* Schleife beendet, ist `continue` deutlich sanfter: Es bricht nur den **aktuellen** Durchlauf ab und springt sofort zurück an den Kopf der Schleife, um die Bedingung dort erneut zu prüfen — die Schleife läuft also ganz normal weiter, nur der Rest des Codes *innerhalb dieses einen Durchlaufs* wird übersprungen.

Man kann sich `continue` wie ein "Überspringen" in einer Playlist vorstellen: Man bricht nicht die gesamte Wiedergabe ab (das wäre `break`), sondern überspringt nur den aktuellen Titel und macht mit dem nächsten normal weiter.

```python
zahl = 0
while zahl < 5:
    zahl = zahl + 1
    if zahl == 3:
        # Wir überspringen die 3
        continue
    print(f"Zahl: {zahl}")
```

Wichtig beim Nachvollziehen: Sobald `zahl` den Wert `3` erreicht, wird `continue` ausgeführt, und die Zeile `print(f"Zahl: {zahl}")` wird für *diesen einen Durchlauf* komplett übersprungen. Die Schleife läuft danach ganz normal weiter zu `zahl = 4` und `zahl = 5`. Das Ergebnis ist eine Ausgabe von `1, 2, 4, 5` — die `3` fehlt gezielt.

**Merkregel:** `break` bedeutet "Schleife komplett verlassen", `continue` bedeutet "diesen einen Durchlauf überspringen, aber weitermachen".

````python
# Live-Coding: continue — einen Wert gezielt überspringen
zahl = 0
while zahl < 5:
    zahl = zahl + 1
    if zahl == 3:
        # Wir überspringen die 3
        continue
    print(f"Zahl: {zahl}")
````

#### Aufgabe 1: Der Patch-Countdown
Schreiben Sie ein Programm, das einen Countdown von 5 bis 1 ausgibt. Nach der 1 soll `"Patchvorgang startet..."` ausgegeben werden.

````python
# Aufgabe 1: Der Patch-Countdown
counter = 5
while counter > 0:
    print(counter)
    counter = counter - 1
print("Patchvorgang startet...")
````

### Kategorie B: Schleifen mit `break` und `continue`

#### Aufgabe 4: Drei Login-Versuche (Sperr-Logik)
Der Benutzer hat maximal 3 Versuche, das Passwort `"geheim"` einzugeben. Wenn er es richtig eingibt, bricht die Schleife mit `break` ab (`"Login erfolgreich"`). Wenn er nach 3 Versuchen scheitert, geben Sie `"Konto gesperrt!"` aus.

````python
# Aufgabe 4: Drei Login-Versuche (Sperr-Logik)
versuche = 1
while versuche <= 3:
    passwort = input(f"Versuch {versuche}/3 - Passwort eingeben: ")
    if passwort == "geheim":
        print("Login erfolgreich!")
        break
    versuche = versuche + 1

if versuche > 3:
    print("Konto gesperrt!")
````

#### Aufgabe 5: Ungerade Ports überspringen
Ein Skript scannt die Ports von `80` bis `86`. Verwenden Sie eine `while`-Schleife und `continue`, um nur die **geraden** Ports auszugeben (Tipp: Nutzen Sie `% 2 == 0` zur Prüfung).

````python
# Aufgabe 5: Ungerade Ports überspringen
port = 79
while port < 86:
    port = port + 1
    if port % 2 != 0:
        continue  # Überspringt ungerade Ports
    print(f"Scanne geraden Port: {port}")
````

#### Aufgabe 6: Der unendliche Chatbot
Schreiben Sie eine Schleife, die den Benutzer immer wieder nach einer Eingabe fragt. Wenn der Benutzer `"exit"`, `"beenden"` oder `"tschüss"` eingibt, soll sich das Programm mit einer Abschiedsmeldung beenden.

````python
# Aufgabe 6: Der unendliche Chatbot
while True:
    eingabe = input("Support-Bot: Wie kann ich helfen? (Typ 'exit' to quit): ").lower()
    if eingabe == "exit" or eingabe == "beenden" or eingabe == "tschüss":
        print("Support-Bot: Sitzung beendet. Auf Wiedersehen!")
        break
    print(f"Support-Bot: Ticket für '{eingabe}' wurde registriert.")
````

### Kategorie C: Komplexe IT-Support Schleifen

#### Aufgabe 7: Server-Überhitzung (Temperatursimulator)
Die Temperatur im Serverraum startet bei `20` Grad.
- In jedem Schleifendurchlauf fragt das Programm den Benutzer: `"Um wie viel Grad steigt die Temperatur? "` (Dezimalzahl).
- Addieren Sie diesen Wert zur Gesamttemperatur.
- Sobald die Temperatur `50` Grad überschreitet, brechen Sie die Schleife ab und geben Sie eine rote Warnmeldung aus: `"NOT-AUSSCHALTUNG AKTIVIERT!"`.

````python
# Aufgabe 7: Server-Überhitzung (Temperatursimulator)
temp = 20.0
while temp <= 50.0:
    print(f"Aktuelle Temperatur: {temp}°C. Zustand stabil.")
    anstieg = float(input("Temperaturanstieg eingeben (°C): "))
    temp = temp + anstieg

print(f"Kritischer Alarm! {temp}°C erreicht. NOT-AUSSCHALTUNG AKTIVIERT!")
````

#### Aufgabe 8: Interaktiver Task-Manager (ToDo-Listen-Schleife)
Schreiben Sie ein interaktives Programm mit einer leeren Liste `offene_tickets = []`.
- Die Schleife fragt den Benutzer fortlaufend nach einer Aktion: `"w"` (Ticket hinzufügen), `"l"` (Tickets anzeigen), `"q"` (Programm beenden).
- Bei `"w"` wird das eingegebene Ticket der Liste hinzugefügt.
- Bei `"l"` wird die aktuelle Liste ausgegeben.
- Bei `"q"` bricht das Programm ab.

````python
# Aufgabe 8: Interaktiver Task-Manager (ToDo-Listen-Schleife)
offene_tickets = []
while True:
    aktion = input("Aktion wählen: [w]rite Ticket, [l]ist Tickets, [q]uit: ").lower()
    if aktion == "q":
        print("Task-Manager beendet.")
        break
    elif aktion == "w":
        ticket = input("Ticket-Beschreibung: ")
        offene_tickets.append(ticket)
        print("Ticket hinzugefügt.")
    elif aktion == "l":
        print(f"Offene Tickets ({len(offene_tickets)}): {offene_tickets}")
    else:
        print("Ungültige Aktion!")
````
