# 📘 Block 1 — Funktionen & Parameter
### 09:15–10:30

## Theorie

**Definition:** Eine Funktion ist ein benannter, wiederverwendbarer Codeblock, der eine klar abgegrenzte Aufgabe erfüllt. Parameter sind Platzhalter für Werte, die beim Aufruf übergeben werden.

**Warum wichtig:** Ohne Funktionen wiederholt sich Logik (z.B. Schaden würfeln) an mehreren Stellen im Code — jede Änderung müsste dann mehrfach gepflegt werden.

**Syntax:**
```python
def funktionsname(parameter1, parameter2):
    """Kurzbeschreibung: Was / Parameter / Rückgabewert"""
    anweisungen
```


````python
def begruesse(name, klasse):
    """
    Gibt eine Begrüßung für einen neuen Helden aus.

    Parameter:
        name (str): Name des Helden
        klasse (str): Heldenklasse, z.B. "Kriegerin"

    Rückgabewert:
        None – gibt direkt auf der Konsole aus.
    """
    print(f"Willkommen, {name} der/die {klasse}!")

begruesse("Aria", "Kriegerin")
````

> 💡 **Merksatz (gilt für den Rest des Kurses):** *„Jede Funktion beantwortet drei Fragen: Was macht sie? Was nimmt sie entgegen? Was gibt sie zurück?"*

# 📘 Block 2 — Rückgabewerte (return)
### 10:45–12:30

## Theorie

**Definition:** `return` beendet eine Funktion sofort und übergibt einen Wert an die aufrufende Stelle, statt ihn nur anzuzeigen.

**Warum wichtig:** `print()` gibt einen Wert nur auf dem Bildschirm aus — danach ist er für das Programm „weg". `return` macht das Ergebnis weiterverarbeitbar, z.B. um es später statt auf der Konsole in einem GUI-Label anzuzeigen.

**Syntax:**
```python
def funktionsname(parameter):
    ...
    return ergebnis
```


````python
import random

def wuerfle_schaden(minimum, maximum):
    """
    Simuliert einen Schadenswurf in einem Bereich.

    Parameter:
        minimum (int): kleinster möglicher Schaden
        maximum (int): größter möglicher Schaden

    Rückgabewert:
        int: der gewürfelte Schaden
    """
    return random.randint(minimum, maximum)

schaden = wuerfle_schaden(5, 20)
print(f"Du fügst {schaden} Schaden zu.")
````

> 💡 **Merksatz:** *„print() zeigt, return liefert. Logik-Funktionen sollten fast immer return nutzen, Anzeige-Funktionen print."*

# 📘 Block 3 — Dateien, Dictionaries & JSON
### 13:30–15:00

## Theorie: Dateien & with

**Definition:** `open()` öffnet eine Datei zum Lesen (`"r"`) oder Schreiben (`"w"`). Der `with`-Block sorgt dafür, dass die Datei automatisch geschlossen wird — auch wenn ein Fehler auftritt.

**Warum wichtig:** Ohne dauerhafte Speicherung geht jeder Spielstand beim Beenden des Programms verloren.

**Syntax:**
```python
with open("pfad", "modus", encoding="utf-8") as datei:
    ...
```


````python
def schreibe_notiz(text, pfad="notiz.txt"):
    """
    Schreibt einen Text in eine Datei (überschreibt vorhandenen Inhalt).

    Parameter:
        text (str): zu speichernder Inhalt
        pfad (str): Zieldatei, Standard "notiz.txt"

    Rückgabewert:
        None
    """
    with open(pfad, "w", encoding="utf-8") as datei:
        datei.write(text)

schreibe_notiz("Aria hat das erste Monster besiegt!")
print("Notiz gespeichert.")
````

> 💡 **Merksatz:** *„Immer `with` statt manuellem open()/close() — sonst riskiert man vergessene, gesperrte Dateien."*

## Theorie: JSON

**Definition:** JSON wandelt ein Python-Dictionary in speicherbaren Text um (`json.dump`) und wieder zurück (`json.load`).

**Warum wichtig:** Ein Charakter besteht aus mehreren zusammengehörigen Werten (Name, HP, Level) — JSON speichert dieses ganze Paket in einer Datei, nicht nur eine einzelne Zahl.

**Syntax:**
```python
json.dump(daten, datei)      # Dictionary -> Datei
geladen = json.load(datei)   # Datei -> Dictionary
```


````python
import json

def speichere_charakter(held, pfad="charakter.json"):
    """
    Speichert den Charakter-Zustand dauerhaft als JSON.

    Parameter:
        held (dict): z.B. {"name": ..., "hp": ..., "level": ...}
        pfad (str): Zieldatei

    Rückgabewert:
        None
    """
    with open(pfad, "w", encoding="utf-8") as datei:
        json.dump(held, datei)

held = {"name": "Aria", "hp": 100, "level": 1}
speichere_charakter(held)
print("Charakter gespeichert.")
````

> 💡 **Merksatz:** *„CSV/reiner Text für einzelne Werte, JSON für zusammengehörige Datenpakete."*

# 🛠️ Block 4 — Praxis: Kampf refactoren & Charakter speichern
### 15:15–17:30

## Gemeinsames Live-Coding (30 Min)

Der Kampf aus Tag 1 wird in eine wiederverwendbare Funktion umgebaut — der erste Schritt weg vom monolithischen Skript.


````python
import random

def kampfrunde(monster_hp, aktion):
    """
    Berechnet den Effekt einer Spieleraktion auf die Monster-HP.

    Parameter:
        monster_hp (int): aktuelle HP des Monsters
        aktion (int): 1 = Angriff, 2 = Verteidigen

    Rückgabewert:
        tuple[int, int]: (neue Monster-HP, verursachter Schaden)
    """
    if aktion == 1:
        schaden = random.randint(5, 20)
        return monster_hp - schaden, schaden
    return monster_hp, 0

monster_hp, schaden = kampfrunde(50, 1)
print(f"Neue Monster-HP: {monster_hp}, Schaden: {schaden}")
````
