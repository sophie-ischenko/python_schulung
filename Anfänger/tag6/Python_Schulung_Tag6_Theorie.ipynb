# Block 2: Bedingungen & Vergleichsoperatoren (Theorie)

## Konzept: Was ist Kontrollfluss?

Bisher lief jedes Skript wie ein stures Rezept ab: Zeile 1 wird ausgeführt, dann Zeile 2, dann Zeile 3 — immer in exakt gleicher Reihenfolge, egal was passiert. Das nennt man **sequenziellen Ablauf**. In der echten Welt trifft aber praktisch jedes IT-System ständig Entscheidungen: "Ist der Speicher voll? Dann sende eine Warnung. Ist er es nicht? Dann mache normal weiter." Genau das bilden wir mit **Bedingungen** (`if`/`elif`/`else`) ab.

**Die Analogie: Die Weggabelung**

Ein Wanderer kommt an eine Weggabelung. Ohne Entscheidungslogik würde er stur immer geradeaus laufen, egal ob dort eine Straße oder ein Abgrund ist. Mit einer `if`-Bedingung kann er sich fragen: "Ist der linke Weg frei? *Wenn ja*, gehe links. *Sonst* gehe rechts." Der `if`-Block markiert also eine Weggabelung im Programmablauf: Je nachdem, ob eine Bedingung wahr (`True`) oder falsch (`False`) ist, nimmt das Programm einen anderen "Pfad" durch den Code.

**Warum ist das der Kern der Programmierung?**

Ohne Bedingungen ist ein Programm nur eine Liste von Befehlen — mit Bedingungen wird es zu einem "intelligenten" System, das auf seine Umgebung reagiert. Ein Monitoring-System, das bei Serverausfall automatisch eine E-Mail verschickt; eine Login-Maske, die nur bei korrektem Passwort Zugang gewährt; eine Backup-Routine, die nur startet, wenn genug Speicherplatz frei ist — all das basiert auf `if`-Bedingungen.

## Syntax: Die Bausteine eines Entscheidungsblocks

1. **Das Schlüsselwort `if`:** leitet die Bedingung ein, bedeutet wörtlich "wenn".
2. **Die Bedingung selbst** (z. B. `speicher < 10`): ein sogenannter **boolescher Ausdruck** — wird von Python immer zu genau einem von zwei möglichen Werten ausgewertet: `True` oder `False`.
3. **Der Doppelpunkt `:`** am Ende der `if`-Zeile ist zwingend erforderlich. Er signalisiert Python: "Jetzt folgt ein Codeblock, der nur unter dieser Bedingung ausgeführt wird."
4. **Die Einrückung (Indentation):** Anders als z. B. in Java oder C++, wo geschweifte Klammern `{}` einen Codeblock markieren, verwendet Python **die Einrückung selbst** als Syntaxelement. Alle Zeilen, die zu einem `if`-Block gehören, müssen um exakt die gleiche Anzahl an Leerzeichen eingerückt sein (üblicherweise 4 Leerzeichen). Fehlt diese Einrückung oder ist sie inkonsistent, wirft Python einen `IndentationError`.
5. **Das Schlüsselwort `else`:** bedeutet "andernfalls" und fängt exakt den Fall ab, in dem die `if`-Bedingung `False` ergeben hat. Ein `else`-Block ist optional.

**Wichtig:** Python prüft die Bedingung genau einmal, an der Stelle, an der der Code ausgeführt wird. Ändert sich die Variable später, wird die `if`-Bedingung nicht automatisch neu geprüft.

```
speicher = 8  # GB freier Speicher

if speicher < 10:
    print("Achtung: Wenig Speicherplatz!")
else:
    print("Speicherplatz ist ausreichend.")
```

## Vergleichsoperatoren — das Vokabular der Bedingungen

| Operator | Bedeutung | Beispiel | Ergebnis |
|---|---|---|---|
| `==` | Gleichheit | `5 == 5` | `True` |
| `!=` | Ungleichheit | `5 != 3` | `True` |
| `<` | Kleiner als | `3 < 5` | `True` |
| `>` | Größer als | `3 > 5` | `False` |
| `<=` | Kleiner oder gleich | `5 <= 5` | `True` |
| `>=` | Größer oder gleich | `4 >= 5` | `False` |

**Die wichtigste Stolperfalle überhaupt:** Der Unterschied zwischen `=` und `==`. Ein einzelnes Gleichheitszeichen `=` ist die **Zuweisung** (`x = 5` bedeutet "speichere die Zahl 5 in der Variable x"). Ein doppeltes Gleichheitszeichen `==` ist der **Vergleichsoperator** — er fragt "sind diese beiden Werte gleich?" und liefert `True` oder `False` zurück, verändert aber keine Variable. Verwechselt man beide in einer `if`-Bedingung, führt das entweder zu einem `SyntaxError` oder — schlimmer noch — zu einem stillen Logikfehler.

```
print(5 == 5)
print(5 != 3)
print(3 < 5)
print(3 > 5)
print(5 <= 5)
print(4 >= 5)
```

# Block 3: Logische Operatoren & Listen-Mitgliedschaften (Theorie)

Eine einzelne Bedingung reicht in der Praxis oft nicht aus. Ein Server ist nicht nur dann kritisch überlastet, wenn die CPU-Auslastung hoch ist — sondern womöglich erst dann, wenn *gleichzeitig* auch der Arbeitsspeicher knapp wird. Für solche Situationen brauchen wir eine Möglichkeit, mehrere Bedingungen **logisch miteinander zu verknüpfen**.

## Logische Operatoren

**`and` (Und-Verknüpfung):** Das Gesamtergebnis ist nur dann `True`, wenn **beide** verknüpften Bedingungen für sich genommen `True` sind. Wie eine Tür mit zwei Schlössern, die beide gleichzeitig aufgeschlossen sein müssen.

**`or` (Oder-Verknüpfung):** Das Gesamtergebnis ist bereits dann `True`, wenn **mindestens eine** der beiden Bedingungen `True` ist. Wie eine Tür mit zwei unabhängigen Klingelknöpfen — es reicht, wenn irgendjemand klingelt.

**`not` (Verneinung):** Dreht einen einzelnen Wahrheitswert um: Aus `True` wird `False`, und aus `False` wird `True`. Wird oft verwendet, um eine Bedingung "umzudrehen" — z. B. `not gesperrt` bedeutet "das System ist *nicht* gesperrt".

**Wahrheitstabelle für `and` und `or`:**

| Bedingung A | Bedingung B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

In der IT-Administration begegnen einem solche Verknüpfungen ständig: "Sende nur dann einen Alarm, wenn die CPU-Auslastung über 80% **und** die RAM-Auslastung über 80% liegt" (beides muss gleichzeitig zutreffen). Oder: "Zeige eine Warnung, wenn die CPU-Auslastung hoch **oder** die RAM-Auslastung hoch ist" (hier reicht bereits eine einzelne Komponente).

```
cpu_auslastung = 85
ram_auslastung = 90

# Beide müssen kritisch sein
if cpu_auslastung > 80 and ram_auslastung > 80:
    print("Kritisch: Gesamter Server ist überlastet!")

# Einer von beiden reicht
if cpu_auslastung > 80 or ram_auslastung > 80:
    print("Warnung: Eine Komponente ist stark ausgelastet.")
```

## Mitgliedschaften prüfen mit `in`

Der `in`-Operator (schon von Tag 4 bei Listen bekannt) prüft, ob ein bestimmter Wert **Mitglied** einer Liste (oder eines Strings) ist, und liefert direkt `True` oder `False` zurück. Kombiniert mit einer `if`-Bedingung lässt sich damit elegant prüfen, ob ein Fall in einer vordefinierten Liste "erlaubter" oder "gesperrter" Werte vorkommt — z. B. bei einer VIP-Kundenliste, einer Blacklist von IP-Adressen oder einer Liste zulässiger Benutzernamen.

Der Vorteil gegenüber dem manuellen Vergleich jedes einzelnen Elements (`kunde == "Müller" or kunde == "Schmidt" or kunde == "Fischer"`) ist die Übersichtlichkeit: Eine einzige `in`-Prüfung gegen eine Liste ersetzt potenziell dutzende einzelne `or`-Verknüpfungen und lässt sich viel leichter erweitern.

```
vip_kunden = ["Müller", "Schmidt", "Fischer"]
kunde = "Schmidt"

if kunde in vip_kunden:
    print("Priorisierter VIP-Support aktivieren.")
else:
    print("Standard-Support-Warteschlange.")
```

## Ausblick

`IndentationError` entsteht, wenn Zeilen innerhalb eines `if`/`elif`/`else`-Blocks nicht konsistent eingerückt sind (z. B. mal mit 4, mal mit 2 Leerzeichen, oder eine Mischung aus Tabs und Leerzeichen).
