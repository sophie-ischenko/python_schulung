# Tag 5: Wochen-Review — Thema 4: Debugging & Fehlersuche

Referenz-Katalog mit 40 typischen Fehlern. Jede Aufgabe zeigt den fehlerhaften Code (als Kommentar) sowie direkt darunter den korrigierten, ausführbaren Code.

### Aufgabe 121: Der String-Klammer-Bug

```
# Fehlerhaft:
# print("Hallo Welt

# Korrektur:
print("Hallo Welt")
```

### Aufgabe 122: Der Variablen-Schreibfehler

```
# Fehlerhaft:
# ip_adresse = "10.0.0.1"
# print(ip_adress)

# Korrektur:
ip_adresse = "10.0.0.1"
print(ip_adresse)
```

### Aufgabe 123: Der Typen-Verknüpfungs-Bug

```
# Fehlerhaft:
# print("Port: " + 80)

# Korrektur:
print("Port: " + "80")  # oder: print(f"Port: {80}")
```

### Aufgabe 124: Der falsche Anführungszeichen-Mix

```
# Fehlerhaft:
# print('Status: "online)

# Korrektur:
print('Status: "online"')
```

### Aufgabe 125: Der unvollständige f-String

```
# Fehlerhaft:
# name = "Webserver"
# print("Server-Name ist: {name}")

# Korrektur:
name = "Webserver"
print(f"Server-Name ist: {name}")
```

### Aufgabe 126: Der Input-Multiplikations-Bug

```
# Fehlerhaft:
# anzahl = input("Menge: ") # Eingabe: 5
# print(anzahl * 2) # Ausgabe ist 55, erwartet wird 10

# Korrektur:
anzahl = int(input("Menge: "))
print(anzahl * 2)
```

### Aufgabe 127: Der vergessene Punkt bei Methoden

```
# Fehlerhaft:
# print("test"upper())

# Korrektur:
print("test".upper())
```

### Aufgabe 128: Der Index-Out-Of-Range-Bug

```
# Fehlerhaft:
# server = ["SRV1", "SRV2"]
# print(server[2])

# Korrektur:
server = ["SRV1", "SRV2"]
print(server[1])
```

### Aufgabe 129: Der Groß-/Kleinschreibungs-Bug bei Boolean

```
# Fehlerhaft:
# aktiv = true

# Korrektur:
aktiv = True
print(aktiv)
```

### Aufgabe 130: Der fehlerhafte Float-Punkt

```
# Fehlerhaft:
# preis = 12,50

# Korrektur:
preis = 12.50
print(preis)
```

### Aufgabe 131: Variable vor Zuweisung genutzt

```
# Fehlerhaft:
# print(status)
# status = "Aktiv"

# Korrektur:
status = "Aktiv"
print(status)
```

### Aufgabe 132: Falsche Klammern bei Print

```
# Fehlerhaft:
# print "Willkommen"

# Korrektur:
print("Willkommen")
```

### Aufgabe 133: Zahl mit führender Null (SyntaxError)

```
# Fehlerhaft:
# port = 080

# Korrektur:
port = 80
print(port)
```

### Aufgabe 134: Vergessenes Komma in Liste

```
# Fehlerhaft:
# ip = ["10.0.0.1" "10.0.0.2"]

# Korrektur:
ip = ["10.0.0.1", "10.0.0.2"]
print(ip)
```

### Aufgabe 135: Addieren von Liste und String

```
# Fehlerhaft:
# liste = ["A", "B"]
# liste = liste + "C"

# Korrektur:
liste = ["A", "B"]
liste.append("C")
print(liste)
```

### Aufgabe 136: Falsches Escape-Zeichen (Pfad)

```
# Fehlerhaft:
# pfad = "C:\temp\newfile.txt"  # Verursacht Fehler wegen \t und \n

# Korrektur:
pfad = "C:\\temp\\newfile.txt"  # oder raw-string: r"C:\temp\newfile.txt"
print(pfad)
```

### Aufgabe 137: Großgeschriebenes Print

```
# Fehlerhaft:
# Print("Start")

# Korrektur:
print("Start")
```

### Aufgabe 138: Falsche Methodenschreibweise

```
# Fehlerhaft:
# print("abc".UPPER())

# Korrektur:
print("abc".upper())
```

### Aufgabe 139: Mathematische Operation auf Text

```
# Fehlerhaft:
# print("10" - 5)

# Korrektur:
print(int("10") - 5)
```

### Aufgabe 140: Fehlerhafte Variablenbenennung

```
# Fehlerhaft:
# 1_server = "SRV-01"

# Korrektur:
server_1 = "SRV-01"
print(server_1)
```

### Aufgabe 141: Der ungültige Konvertierungs-Bug

```
# Fehlerhaft:
# wert = int("fünf")

# Korrektur:
wert = int("5")
print(wert)
```

### Aufgabe 142: Der Division-durch-Null-Bug

```
# Fehlerhaft:
# last, server = 100, 0
# schnitt = last / server

# Korrektur:
last, server = 100, 1
schnitt = last / server
print(schnitt)
```

### Aufgabe 143: Der String-Subtraktions-Bug

```
# Fehlerhaft:
# text = "ID-99"
# zahl = text - "ID-"

# Korrektur:
text = "ID-99"
zahl = text.replace("ID-", "")
print(zahl)
```

### Aufgabe 144: Der falsche Klammer-Typ bei Listen

```
# Fehlerhaft:
# daten = (1, 2, 3)  # Tuple statt Liste
# daten.append(4)

# Korrektur:
daten = [1, 2, 3]
daten.append(4)
print(daten)
```

### Aufgabe 145: Der fehlende Wert bei .remove()

```
# Fehlerhaft:
# daten = ["A", "B"]
# daten.remove()

# Korrektur:
daten = ["A", "B"]
daten.remove("A")
print(daten)
```

### Aufgabe 146: Der Doppel-Input-Klammer-Bug

```
# Fehlerhaft:
# anzahl = int(input("Anzahl: "

# Korrektur:
anzahl = int(input("Anzahl: "))
print(anzahl)
```

### Aufgabe 147: Der fehlerhafte List-Index-Zugriff

```
# Fehlerhaft:
# daten = [10, 20, 30]
# print(daten("1"))

# Korrektur:
daten = [10, 20, 30]
print(daten[1])
```

### Aufgabe 148: Der falsche f-String-Inhalt

```
# Fehlerhaft:
# rechner = "Client-01"
# print(f"Rechner ist rechner")

# Korrektur:
rechner = "Client-01"
print(f"Rechner ist {rechner}")
```

### Aufgabe 149: Der None-Rückgabe-Bug bei .sort()

```
# Fehlerhaft:
# liste = [3, 1, 2]
# sortierte_liste = liste.sort()
# print(sortierte_liste)  # Gibt None aus

# Korrektur:
liste = [3, 1, 2]
liste.sort()
print(liste)
```

### Aufgabe 150: Der unerwartete Einrückungs-Bug

```
# Fehlerhaft:
# print("Schritt 1")
#   print("Schritt 2")

# Korrektur:
print("Schritt 1")
print("Schritt 2")
```

### Aufgabe 151: Fehlerhaftes Format-Casting beim Input

```
# Fehlerhaft:
# preis = int(input("Preis: "))  # User gibt 19.99 ein -> Absturz
# print(preis)

# Korrektur:
preis = float(input("Preis: "))
print(preis)
```

### Aufgabe 152: Löschen von nicht existenten Listenelementen

```
# Fehlerhaft:
# users = ["Tim", "Jan"]
# users.remove("Ben")  # Absturz, da "Ben" nicht existiert

# Korrektur:
users = ["Tim", "Jan"]
if "Ben" in users:
    users.remove("Ben")
print(users)
```

### Aufgabe 153: Ungültige Methode für Datentyp

```
# Fehlerhaft:
# ip = 192.168
# print(ip.split("."))  # Absturz, da Float kein split() besitzt

# Korrektur:
ip = "192.168"
print(ip.split("."))
```

### Aufgabe 154: Gefangene Eingabe im f-String

```
# Fehlerhaft:
# print(f"Die Summe ist: {10 + "20"}")

# Korrektur:
print(f"Die Summe ist: {10 + int('20')}")
```

### Aufgabe 155: Index-Suche fehlgeschlagen

```
# Fehlerhaft:
# namen = ["Alex", "Ben"]
# idx = namen.index("Chris")  # Absturz, wenn "Chris" nicht da ist

# Korrektur:
namen = ["Alex", "Ben"]
idx = namen.index("Chris") if "Chris" in namen else -1
print(idx)
```

### Aufgabe 156: Zuweisungs-Gleichheits-Verwechslung

```
# Fehlerhaft:
# # Wir wollen prüfen, ob der Status "Aktiv" ist (Fehler bei der Zuweisung)
# status = "Inaktiv"
# status = "Aktiv"  # Sollte Prüfung sein

# Korrektur:
status = "Inaktiv"
print(status == "Aktiv")  # True/False Prüfung
```

### Aufgabe 157: Unlesbares Listen-Slicing

```
# Fehlerhaft:
# daten = [10, 20, 30]
# print(daten[1, 2])  # Falsche Syntax für Slicing

# Korrektur:
daten = [10, 20, 30]
print(daten[1:3])
```

### Aufgabe 158: Doppelter Operator-Fehler

```
# Fehlerhaft:
# rechnung = 10 ++ 5  # Kein direkter Absturz, aber logisch unschön

# Korrektur:
rechnung = 10 + 5
print(rechnung)
```

### Aufgabe 159: Falsche Syntax für String-Multiplikation

```
# Fehlerhaft:
# print("A" x 5)

# Korrektur:
print("A" * 5)
```

### Aufgabe 160: Nicht-String in .lower() Aufruf

```
# Fehlerhaft:
# code = 1024
# print(code.lower())  # Absturz, da Integer keine .lower()-Methode hat

# Korrektur:
code = "1024"
print(code.lower())
```
