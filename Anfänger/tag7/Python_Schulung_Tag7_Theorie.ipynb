Sechs gezielte Debugging-Fälle zu `if`/`else` und Operatoren: jeweils fehlerhafter Code mit Diagnose, danach die reparierte Version.

## Debug-Aufgabe D1: Die verschobene Einrückung (IndentationError)

**Diagnose:** Python definiert Code-Blöcke strikt über Einrückungen. Die Befehle nach dem `if:` müssen genau um dieselbe Anzahl an Leerzeichen (üblicherweise 4) eingerückt sein und auf derselben Ebene liegen. Im fehlerhaften Beispiel ist die erste `print`-Zeile gar nicht eingerückt (obwohl sie zum `if`-Block gehören soll) und die zweite ist inkonsistent eingerückt — beides erzeugt einen `IndentationError`.

*Der fehlerhafte Code ist auskommentiert, da ein `IndentationError` sonst die ganze Zelle blockiert. Zum Ausprobieren die Kommentarzeichen entfernen.*

````
# Fehlerhafter Code (auskommentiert)
# speicher_voll = True
#
# if speicher_voll:
# print("Warnung: Speicher voll!")
#   print("Bitte Backup starten.")

# Reparierter Code:
speicher_voll = True
if speicher_voll:
    print("Warnung: Speicher voll!")
    print("Bitte Backup starten.")
````

## Debug-Aufgabe D2: Die vergessenen Doppelpunkte (SyntaxError)

**Diagnose:** Nach einer `if`- oder `else`-Bedingung muss zwingend ein Doppelpunkt `:` stehen, um den nachfolgenden Block einzuleiten. Im Beispiel fehlt der Doppelpunkt sowohl nach der `if`- als auch nach der `else`-Zeile.

````
# Fehlerhafter Code (auskommentiert)
# status = "offline"
#
# if status == "offline"
#     print("Server ist offline.")
# else
#     print("Server läuft.")

# Reparierter Code:
status = "offline"
if status == "offline":
    print("Server ist offline.")
else:
    print("Server läuft.")
````

## Debug-Aufgabe D3: Die Zuweisung in der Bedingung (SyntaxError)

**Diagnose:** Ein einzelnes Gleichheitszeichen `=` ist eine Zuweisung. Für Vergleiche in Bedingungen muss zwingend das doppelte Gleichheitszeichen `==` verwendet werden. `if port = 80:` ist deshalb kein gültiger Vergleich, sondern ein `SyntaxError` — Python verhindert genau diese Verwechslung.

````
# Fehlerhafter Code (auskommentiert)
# port = 80
#
# if port = 80:
#     print("Webserver aktiv.")

# Reparierter Code:
port = 80
if port == 80:
    print("Webserver aktiv.")
````

## Debug-Aufgabe D4: Logische Verwechslung (Logischer Fehler)

*Ziel:* Der Zugriff ist erlaubt, wenn der Benutzer `"Admin"` ODER `"Superuser"` heißt.

**Diagnose:** Dieser Code wirft keinen Fehler und lässt sich normal ausführen — das macht ihn besonders tückisch, denn das Programm liefert das *falsche* Ergebnis. Mit `and` verlangt Python, dass der Benutzername *gleichzeitig* `"Admin"` **und** `"Superuser"` ist — unmöglich, da eine Variable nicht gleichzeitig zwei Werte haben kann. Die Bedingung ist also immer `False`. Hier muss `or` verwendet werden.

````
# Fehlerhafter Code (läuft, liefert aber falsches Ergebnis)
username = "Superuser"

if username == "Admin" and username == "Superuser":
    print("Zugriff erlaubt.")
else:
    print("Zugriff verweigert.")
````

````
# Reparierter Code
username = "Superuser"
if username == "Admin" or username == "Superuser":
    print("Zugriff erlaubt.")
else:
    print("Zugriff verweigert.")
````

## Debug-Aufgabe D5: Der unberechenbare String (TypeError)

**Diagnose:** `input()` liefert *immer* einen String zurück, selbst wenn der Benutzer eine Zahl eingibt. Der Vergleich `ping < 50` versucht, Text mit einer Zahl zu vergleichen — das führt zu einem `TypeError`. Die Eingabe muss zuerst mit `int()` konvertiert werden.

````
# Fehlerhafter Code — bitte mit Eingabe "45" ausführen
ping = input("Geben Sie den Ping in ms ein: ")  # User gibt 45 ein

if ping < 50:
    print("Verbindung ist schnell.")
````

````
# Reparierter Code
ping = int(input("Geben Sie den Ping in ms ein: "))
if ping < 50:
    print("Verbindung ist schnell.")
````

## Debug-Aufgabe D6: Doppelte Verneinung (Logischer Fehler)

*Ziel:* Der Server soll patchen, wenn er *nicht* gesperrt ist (`gesperrt = False`).

**Diagnose:** Auch dieser Code läuft ohne Absturz durch, liefert aber ein verwirrendes/falsches Ergebnis, weil er den Wahrheitswert doppelt verneint: Wenn `gesperrt` `False` ist, dann ist `gesperrt == False` bereits `True`. Das zusätzliche `not` davor dreht dieses `True` wieder zu `False` um — genau das Gegenteil vom Gewollten. `not gesperrt` reicht vollkommen aus.

````
# Fehlerhafter Code (läuft, aber mit falscher/verwirrender Logik)
gesperrt = False

if not gesperrt == False:
    print("Patch-Vorgang startet.")
else:
    print("Server ist gesperrt!")
````

````
# Reparierter Code
gesperrt = False
if not gesperrt:
    print("Patch-Vorgang startet.")
else:
    print("Server ist gesperrt!")
````

# Block 2: Einführung in die while-Schleife (Theorie)

## Das Konzept von Schleifen — warum wir sie brauchen

Bisher musste man Code von Hand kopieren, wenn man etwas mehrmals ausführen wollte. Für viermal "Ping gesendet" hätte man die `print`-Zeile viermal abtippen müssen. Das funktioniert bei 4 Wiederholungen noch, aber was, wenn es 400 oder 40.000 sein sollen — oder wenn die genaue Anzahl vorher gar nicht bekannt ist (z. B. "frage so lange nach dem Passwort, bis es stimmt")? Genau dafür gibt es **Schleifen (Loops)**.

**Die Analogie: Der Wachmann an der Schranke**

Ein Wachmann an einer Parkplatzschranke hat eine klare Regel: "Solange noch Autos in der Schlange warten, lasse das nächste durch." Er prüft diese Bedingung nicht nur einmal, sondern **immer wieder, vor jedem einzelnen Auto neu** — und erst wenn die Bedingung nicht mehr zutrifft, hört er auf. Genau so funktioniert eine `while`-Schleife: Sie prüft vor *jedem einzelnen* Durchlauf erneut, ob die Bedingung noch wahr ist, und wiederholt den Codeblock so lange, wie das der Fall ist.

## Syntax der while-Schleife

Durchlauf für Durchlauf mit `zaehler` startend bei `1` und Bedingung `zaehler <= 3`:
- **1. Prüfung:** `1 <= 3` → `True`. Ausgabe "Durchlauf Nummer: 1", `zaehler` wird auf `2` erhöht.
- **2. Prüfung:** `2 <= 3` → `True`. Ausgabe "Durchlauf Nummer: 2", `zaehler` wird auf `3` erhöht.
- **3. Prüfung:** `3 <= 3` → `True`. Ausgabe "Durchlauf Nummer: 3", `zaehler` wird auf `4` erhöht.
- **4. Prüfung:** `4 <= 3` → `False`. Die Schleife wird beendet.

**Der entscheidende Baustein:** die Zeile `zaehler = zaehler + 1`. Ohne diese Zeile würde sich `zaehler` niemals verändern, die Bedingung wäre also *für immer* `True`, und die Schleife würde nie enden. Diese Zeile ist der Mechanismus, der überhaupt dafür sorgt, dass die Schleife irgendwann aufhört.

````
zaehler = 1

while zaehler <= 3:
    print(f"Durchlauf Nummer: {zaehler}")
    zaehler = zaehler + 1  # Wichtig: Zähler erhöhen!
````

## ⚡ Achtung: Die Endlosschleife (Unendlicher Loop)

Wenn die Bedingung der Schleife *niemals* `False` wird — sei es, weil man vergisst, den Zähler zu erhöhen, oder weil man absichtlich `while True:` schreibt — läuft das Programm unendlich weiter, bis man es von außen stoppt. In einer echten Anwendung kann das den Arbeitsspeicher füllen oder das gesamte Programm einfrieren lassen.

*Praxis-Hinweis:* In Google Colab (oder Jupyter) klickt man in so einem Fall sofort auf den **Stop-Button ("Interrupt execution")** links neben der Zelle, bzw. im Menü auf "Laufzeit" → "Ausführung unterbrechen". Eine Endlosschleife lässt sich jederzeit von außen abbrechen, sie ist kein dauerhafter Schaden am System, sondern lediglich eine laufende Berechnung, die man stoppen muss.

`while True:` selbst ist kein Fehler, sondern ein bewusst eingesetztes, sehr gebräuchliches Muster — nämlich immer dann, wenn wir die Schleife absichtlich von *innen* heraus beenden wollen, statt über eine Bedingung am Anfang. Genau das folgt jetzt in Block 3 mit `break`.

# Block 3: Loop-Steuerung mit break und continue (Theorie)

## Schleife abbrechen mit break

Mit `break` beenden wir die Schleife **sofort und vollständig** — egal, ob die eigentliche Schleifen-Bedingung noch wahr wäre oder nicht. Wie ein Notausschalter: Sobald `break` ausgeführt wird, springt das Programm direkt aus der Schleife heraus und macht mit dem Code danach weiter, ohne die Bedingung noch einmal zu prüfen.

Besonders nützlich in Kombination mit `while True:` — einer Schleife, deren Kopf-Bedingung *immer* wahr ist und die sich daher niemals von selbst beendet. Solche Schleifen wirken wie Endlosschleifen, sind es aber nicht, weil irgendwo *im Inneren* der Schleife ein `break` versteckt ist, das unter der richtigen Bedingung ausgelöst wird. Extrem verbreitet bei Menüs, Chatbots oder Login-Systemen, bei denen man vorher nicht weiß, wie oft die Schleife durchlaufen werden muss — nur, *unter welcher Bedingung* sie enden soll.

````
# Das unendliche Login
while True:
    eingabe = input("Passwort eingeben (oder 'exit' zum Abbrechen): ")
    if eingabe == "admin123" or eingabe == "exit":
        print("Schleife wird beendet.")
        break  # Springt sofort aus der Schleife
    print("Falsches Passwort! Versuchen Sie es erneut.")
````

## Schritte überspringen mit continue

Während `break` die *gesamte* Schleife beendet, ist `continue` sanfter: Es bricht nur den **aktuellen** Durchlauf ab und springt sofort zurück an den Kopf der Schleife, um die Bedingung erneut zu prüfen — die Schleife läuft normal weiter, nur der Rest des Codes *innerhalb dieses einen Durchlaufs* wird übersprungen.

Wie ein "Überspringen" in einer Playlist: Man bricht nicht die gesamte Wiedergabe ab (das wäre `break`), sondern überspringt nur den aktuellen Titel und macht mit dem nächsten weiter.

Im Beispiel unten: Sobald `zahl` den Wert `3` erreicht, wird `continue` ausgeführt, und die `print`-Zeile wird für *diesen einen Durchlauf* übersprungen. Die Schleife läuft danach normal weiter zu `zahl = 4` und `zahl = 5`. Ergebnis: Ausgabe von `1, 2, 4, 5` — die `3` fehlt gezielt.

**Merkregel:** `break` bedeutet "Schleife komplett verlassen", `continue` bedeutet "diesen einen Durchlauf überspringen, aber weitermachen".

````
zahl = 0
while zahl < 5:
    zahl = zahl + 1
    if zahl == 3:
        # Wir überspringen die 3
        continue
    print(f"Zahl: {zahl}")
````
