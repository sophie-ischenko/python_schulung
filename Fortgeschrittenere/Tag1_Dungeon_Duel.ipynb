# 📘 Block 1 — Variablen & Datentypen

## Theorie

**Definition:** Eine Variable ist ein benannter Speicherplatz für einen Wert. Python erkennt den Datentyp automatisch anhand des zugewiesenen Werts (dynamische Typisierung).

**Warum wichtig:** Jede Berechnung und jeder Vergleich im Spiel setzt voraus, dass wir wissen, mit welchem Datentyp wir es zu tun haben — besonders, weil `input()` immer Text liefert, selbst wenn Ziffern eingegeben werden.

**Syntax:**
```python
variablenname = wert
```


````python
held_name = "Aria"          # str    – Text
held_hp = 100                # int    – Ganzzahl
kritische_chance = 0.15      # float  – Kommazahl
ist_betaeubt = False         # bool   – Wahrheitswert

# input() liefert IMMER einen String:
eingabe = input("Wähle 1 oder 2: ")   # eingabe ist "1", nicht 1
aktion = int(eingabe)                  # jetzt erst eine echte Zahl
print(type(eingabe), type(aktion))
````

# 📘 Block 2 — Vergleiche & if/elif/else
### 10:45–12:30

## Theorie

**Definition:** Vergleichsoperatoren (`>`, `<`, `==`, `!=`, `>=`, `<=`) liefern immer `True` oder `False`. `if`/`elif`/`else` steuert, welcher Code abhängig davon ausgeführt wird.

**Warum wichtig:** Jede Entscheidung im Spiel — trifft ein Angriff, ist ein Charakter besiegt, war die Eingabe gültig — läuft über diese Bausteine.

**Syntax:**
```python
if bedingung1:
    ...
elif bedingung2:
    ...
else:
    ...
```


````python
def werte_angriff(wurf, ruestung):
    """
    Bewertet einen Angriffswurf gegen die Rüstung des Ziels.

    Parameter:
        wurf (int): simulierter Würfelwurf, 1-20
        ruestung (int): Rüstungswert des Ziels

    Rückgabewert:
        str: "treffer", "patzer" oder "verfehlt"
    """
    if wurf == 1:
        return "patzer"
    elif wurf >= ruestung:
        return "treffer"
    else:
        return "verfehlt"

print(werte_angriff(1, 10))
print(werte_angriff(15, 10))
print(werte_angriff(5, 10))
````

> 💡 **Merksatz:** *„Ein `=` speichert, zwei `==` vergleichen. Einrückung ist in Python der Code-Block — keine Klammern nötig."*

# 📘 Block 3 — Die while-Schleife
### 13:30–14:30

## Theorie

**Definition:** Eine `while`-Schleife wiederholt einen Codeblock, solange eine Bedingung `True` ist. `while True:` in Kombination mit `break` wird genutzt, wenn die Abbruchbedingung erst *innerhalb* der Schleife entsteht.

**Warum wichtig:** Ein Kampf dauert eine unbekannte Anzahl Runden — wir wissen vorher nicht, wie oft die Schleife laufen muss, bis eine Seite besiegt ist.

**Syntax:**
```python
while bedingung:
    ...

while True:
    ...
    if abbruchbedingung:
        break
```


````python
monster_hp = 50
runde = 0

while monster_hp > 0:
    runde += 1
    monster_hp -= 10   # Kurzschreibweise für monster_hp = monster_hp - 10

print(f"Besiegt nach {runde} Runden")
````

> 💡 **Merksatz:** *„Vergisst man, die Bedingung in der Schleife zu verändern, entsteht eine Endlosschleife — ein Klassiker, den man einmal bewusst erlebt haben sollte."*

# 📘 Block 4 — Fehlerbehandlung (try/except)
### 14:45–15:45

## Theorie

**Definition:** `try/except` fängt einen definierten Fehlertyp ab, der beim Ausführen von Code auftreten kann, statt das Programm abstürzen zu lassen.

**Warum wichtig:** Benutzereingaben sind nie vollständig kontrollierbar. Ohne Absicherung stürzt das Programm bei jeder unerwarteten Eingabe ab.

**Syntax:**
```python
try:
    riskanter_code
except FehlerTyp:
    reaktion_auf_fehler
```


````python
while True:
    eingabe = input("Aktion (1=Angriff, 2=Verteidigen): ")
    try:
        aktion = int(eingabe)
    except ValueError:
        print("Bitte nur 1 oder 2 eingeben!")
        continue   # springt zurück an den Schleifenanfang
    break

print("Gewählte Aktion:", aktion)
````

> 💡 **Merksatz:** *„Gezielt `except ValueError` statt pauschal `except Exception` — sonst verschluckt man auch Fehler, die man eigentlich bemerken sollte."*

# 🛠️ Block 5 — Praxis: Den Kampf zusammenbauen
### 16:00–17:30

## Gemeinsames Live-Coding (30 Min)

Das ist der Startpunkt für **Dungeon Duel** — dieses Skript wird die ganze Woche über Schritt für Schritt ausgebaut.


````python
import random

INTRO_TEXT = """
--------------------------------------------------
⚔️  Willkommen bei DUNGEON DUEL  ⚔️
Ein wildes Monster erscheint! Kämpfe um dein Leben!
--------------------------------------------------
"""

print(INTRO_TEXT)
held_name = input("Wie heißt dein Held? ")
print(f"Viel Erfolg im Kampf, {held_name}!\n")

held_hp = 100
monster_hp = 50
runde = 0

while True:
    runde += 1
    eingabe_text = input("Aktion wählen (1 = Angriff, 2 = Verteidigen): ")

    try:
        aktion = int(eingabe_text)
    except ValueError:
        print("Fehler: Bitte nur 1 oder 2 eingeben!")
        continue

    if aktion == 1:
        schaden = random.randint(5, 20)
        monster_hp -= schaden
        print(f"Du greifst an und fügst {schaden} Schaden zu! (Monster HP: {max(monster_hp, 0)})")
    elif aktion == 2:
        print("Du gehst in Verteidigungsstellung.")
    else:
        print("Ungültige Aktion! Wähle 1 oder 2.")
        continue

    if monster_hp <= 0:
        print(f"🎉 🎉 🎉 DU HAST DAS MONSTER BESIEGT! 🎉 🎉 🎉")
        print(f"Du hast {runde} Runden gebraucht.")
        break

    monster_schaden = random.randint(5, 15)
    if aktion == 2:
        monster_schaden = monster_schaden // 2   # Verteidigen halbiert den Schaden
    held_hp -= monster_schaden
    print(f"Das Monster greift an und fügt dir {monster_schaden} Schaden zu! (Deine HP: {max(held_hp, 0)})")

    if held_hp <= 0:
        print("💀 Du wurdest besiegt... 💀")
        break
````
