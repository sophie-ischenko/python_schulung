## 10:00 – 10:45 Uhr | Block 2: Einführung in Listen (Theorie)

Nun folgt der neue Lerninhalt des Tages — und er ist einer der wichtigsten Bausteine überhaupt, denn fast jedes reale Programm muss irgendwann *mehrere* Dinge gleichzeitig verwalten.

### 1. Was ist eine Liste? — Das Konzept ganz von vorne

Bisher konnte eine Variable immer nur **einen einzigen Wert** speichern. Wenn Sie `hostname = "SRV-01"` geschrieben haben, dann passte in diese Variable genau ein Servername — und wenn Sie den zweiten Server auch speichern wollten, brauchten Sie eine zweite Variable (`hostname2 = "SRV-02"`), dann eine dritte, und so weiter. Das funktioniert bei 2 oder 3 Servern noch halbwegs, aber was ist, wenn Sie 50, 500 oder 5000 Server verwalten müssen? Mit einzelnen Variablen wäre das komplett unpraktikabel.

**Die Analogie: Der Aktenschrank**
Stellen Sie sich eine Liste wie einen Aktenschrank mit nummerierten Schubladen vor. Anstatt für jedes Dokument einen eigenen, frei herumstehenden Tisch zu haben (= eine eigene Variable), packen Sie alle zusammengehörigen Dokumente in *einen* Schrank. Jede Schublade hat eine Nummer, und Sie können gezielt "Schublade 3" öffnen, ohne die anderen anfassen zu müssen. Der große Vorteil: Der Schrank selbst hat nur *einen* Namen (z. B. `server_liste`), aber er kann beliebig viele Werte aufnehmen — und Sie können später jederzeit neue Schubladen hinzufügen oder alte entfernen.

**Warum ist das in der IT so wichtig?**
In der Praxis begegnen Ihnen Listen ständig: eine Liste aller aktiven Benutzer, eine Liste aller offenen Tickets, eine Liste aller IP-Adressen in einem Netzwerk, eine Liste aller zu installierenden Programme. Ohne Listen (und die Schleifen, die wir morgen kennenlernen) müssten Sie für jeden einzelnen Server, jeden einzelnen Benutzer eigenen Code schreiben. Mit Listen können Sie stattdessen **eine Datenstruktur** definieren, die beliebig wächst oder schrumpft.

**Syntax:** Eckige Klammern `[]` markieren den Anfang und das Ende der Liste. Die einzelnen Werte (auch "Elemente" genannt) werden durch Kommas voneinander getrennt.
```python
server_liste = ["SRV-01", "SRV-02", "SRV-03"]
```
Wichtig zu wissen: Eine Liste kann grundsätzlich beliebige Datentypen enthalten — auch gemischt. `["SRV-01", 42, 3.14, True]` ist technisch eine gültige Liste, auch wenn man in der Praxis meist versucht, in einer Liste möglichst "gleichartige" Dinge zu sammeln (z. B. nur Servernamen, oder nur Zahlen), damit der Code später übersichtlich bleibt.

### 2. Zugriff über den Index — Wie man eine einzelne Schublade öffnet

Genau wie bei Strings (die Sie an Tag 3 kennengelernt haben) hat jedes Element in einer Liste eine feste, durchnummerierte Position, einen sogenannten **Index**. Und genau wie bei Strings beginnt diese Nummerierung **bei 0, nicht bei 1**. Das ist zu Beginn oft die größte Stolperfalle für Einsteiger: Das *erste* Element einer Liste hat den Index `0`, das *zweite* Element hat den Index `1`, und so weiter.

```text
 Liste:     "192.168.1.1"   "192.168.1.2"   "192.168.1.3"
 Index+:          0               1               2
 Index-:         -3              -2              -1
```

Genau wie bei Strings funktionieren auch **negative Indizes**: `ips[-1]` liefert immer das *letzte* Element, egal wie lang die Liste ist — sehr praktisch, wenn Sie z. B. den zuletzt hinzugefügten Log-Eintrag abrufen wollen, ohne vorher zählen zu müssen, wie viele Einträge insgesamt vorhanden sind.

Eine weitere sehr nützliche eingebaute Funktion ist `len()` (kurz für "length" = Länge). Sie verrät Ihnen, wie viele Elemente aktuell in der Liste enthalten sind — das ist besonders wichtig, weil sich die Länge einer Liste ja jederzeit ändern kann (dazu gleich mehr in Block 3).

**Gemeinsames Live-Coding:**

````python
# Live-Coding: Zugriff über den Index
ips = ["192.168.1.1", "192.168.1.2", "192.168.1.3"]
print(ips[0])   # Erstes Element
print(ips[-1])  # Letztes Element
print(len(ips)) # Anzahl der Elemente
````

### 3. Elemente in einer Liste ändern — der entscheidende Unterschied zu Strings

Hier kommt ein zentraler konzeptioneller Unterschied, den man sich unbedingt merken sollte: **Strings sind unveränderlich (immutable)**, das heißt, ein einzelnes Zeichen in einem bestehenden String kann man *nicht* nachträglich austauschen — man kann nur einen komplett neuen String erzeugen (so wie wir es bei `.replace()` gesehen haben). **Listen dagegen sind veränderlich (mutable)** — das bedeutet, wir können ein einzelnes Element direkt an seiner Position im Speicher überschreiben, ohne die ganze Liste neu zu erstellen.

Das ist so, als würden Sie in unserem Aktenschrank-Bild einfach den Inhalt von Schublade Nummer 2 austauschen, ohne den gesamten Schrank neu zu bauen: Die Adresse "Schublade 2" bleibt gleich, nur der Inhalt ändert sich.

```python
hardware = ["Maus", "Tastatur", "Monitor"]
hardware[1] = "Headset"
print(hardware)  # ['Maus', 'Headset', 'Monitor']
```

Diese Eigenschaft macht Listen enorm flexibel für alles, was sich im Laufe eines Programms verändert — zum Beispiel ein IT-Inventar, dessen Bestand sich ständig ändert, oder eine Warteschlange von Support-Tickets, die abgearbeitet werden.

````python
# Live-Coding: Elemente in einer Liste ändern
hardware = ["Maus", "Tastatur", "Monitor"]
hardware[1] = "Headset"
print(hardware)
````

## 11:00 – 11:30 Uhr | Block 3: List-Methoden & dynamische Veränderung (Theorie)

Bisher hatten unsere Listen immer eine feste Größe. In der Praxis ist das jedoch selten der Fall: Neue Server kommen hinzu, alte werden abgebaut; neue Tickets treffen ein, bearbeitete werden geschlossen. Python stellt uns dafür eingebaute **Methoden** zur Verfügung, mit denen wir Listen dynamisch, also zur Laufzeit, wachsen und schrumpfen lassen können.

### 1. Elemente hinzufügen — die Liste wachsen lassen

**`.append(wert)`** ist die mit Abstand am häufigsten verwendete Listen-Methode. Sie hängt einen neuen Wert ganz **hinten** an die bestehende Liste an — so, als würden Sie im Aktenschrank einfach eine weitere Schublade unten anbauen. Die Reihenfolge der bereits vorhandenen Elemente bleibt dabei unverändert; es kommt lediglich ein neues Element am Ende hinzu.

**`.insert(index, wert)`** ist die präzisere, aber seltener benötigte Variante: Sie erlaubt es, ein neues Element an einer **bestimmten Position** einzuschieben, statt es einfach nur ans Ende zu hängen. Alle Elemente ab dieser Position rutschen dabei automatisch eine Position weiter nach hinten — Python kümmert sich intern um diese Verschiebung, Sie müssen sich darum nicht kümmern.

Der Unterschied lässt sich mit einer Warteschlange am Bahnschalter vergleichen: `.append()` ist wie ein neuer Kunde, der sich hinten anstellt. `.insert(0, ...)` ist wie ein Kunde, der (mit Erlaubnis) ganz vorne reingelassen wird, weil sein Anliegen dringend ist — alle anderen rücken automatisch eine Position nach hinten.

### 2. Elemente entfernen — die Liste schrumpfen lassen

**`.remove(wert)`** durchsucht die Liste nach dem angegebenen **Wert** (nicht nach einer Position!) und löscht das *erste* Vorkommen davon. Wichtig: Wenn der Wert mehrfach in der Liste vorkommt, wird nur die erste Fundstelle entfernt. Und wenn der Wert überhaupt nicht in der Liste enthalten ist, wirft Python einen `ValueError` — das ist ein häufiger Debugging-Fall in der Praxis.

**`.pop(index)`** funktioniert umgekehrt: Sie geben eine **Position** an (nicht einen Wert), und Python entfernt das Element an genau dieser Stelle. Der entscheidende Zusatznutzen von `.pop()`: Es **gibt den entfernten Wert zurück**, sodass man ihn direkt in einer neuen Variablen weiterverarbeiten kann — sehr nützlich z. B. bei einer Ticket-Warteschlange, wenn man wissen möchte, *welches* Ticket gerade aus der Warteschlange geholt und zur Bearbeitung übergeben wurde. Ruft man `.pop()` ganz ohne Angabe eines Index auf, wird standardmäßig das *letzte* Element entfernt.

Zur Merkhilfe: **`.remove()`** = "ich weiß, *was* weg soll, aber nicht wo es steht" — **`.pop()`** = "ich weiß, *wo* es steht (die Position), und ich möchte den entfernten Wert auch noch weiterverwenden können".

**Gemeinsames Live-Coding:**

````python
# Live-Coding: List-Methoden im Zusammenspiel
tools = ["Ping", "Nslookup"]
tools.append("Tracert")
tools.insert(1, "Nmap")
print("Nach Hinzufügen:", tools)

tools.remove("Ping")
entfernt = tools.pop(0)
print("Verbleibend:", tools)
print("Entfernt wurde:", entfernt)
````
