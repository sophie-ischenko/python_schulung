## Variablen

Eine Variable ist vergleichbar mit einem beschrifteten Umzugskarton des Computers. Wir legen einen Wert hinein und schreiben einen Namen darauf.

```
alter = 25
name = "Sabine"

print(alter)
print(name)
```

### Typische Stolperfallen bei Variablen

**1. ❌ Variablennamen nicht in Anführungszeichen setzen**

```
# Falsch:
# "abteilung" = "IT-Support"

# Richtig:
abteilung = "IT-Support"
print(abteilung)
```

**2. ❌ Strings müssen in Anführungszeichen stehen**

Ohne Anführungszeichen sucht Python nach einer Variable und interpretiert `-` als Minus.

```
# Falsch:
# abteilung = IT-Support

# Richtig:
abteilung = "IT-Support"
print(abteilung)
```

**3. ❌ Groß- und Kleinschreibung beachten**

`abteilung` und `Abteilung` sind zwei verschiedene Variablennamen.

```
abteilung = "IT-Support"
print(abteilung)   # funktioniert
# print(Abteilung)   # würde einen NameError auslösen
```

**4. ❌ Leerzeichen im Variablennamen sind nicht erlaubt**

```
# Falsch:
# meine abteilung = "IT-Support"

# Richtig:
meine_abteilung = "IT-Support"
print(meine_abteilung)
```

**5. ❌ Keine Bindestriche im Variablennamen**

Der Bindestrich wird als Minus interpretiert.

```
# Falsch:
# it-support = "IT-Support"

# Richtig:
it_support = "IT-Support"
print(it_support)
```

**6. ❌ Variablen müssen vor der Verwendung existieren**

```
# Falsch (in dieser Reihenfolge):
# print(abteilung)
# abteilung = "IT-Support"

# Richtig:
abteilung = "IT-Support"
print(abteilung)
```

**7. ❌ Reservierte Wörter dürfen nicht als Variablennamen verwendet werden**

z. B. `class` ist reserviert.

```
# Falsch:
# class = "IT-Support"

# Richtig:
abteilung = "IT-Support"
```

## Einfache vs. doppelte Anführungszeichen

In Python sind einfache (`' '`) und doppelte (`" "`) Anführungszeichen grundsätzlich gleichwertig.

```
abteilung = "IT-Support"
print(abteilung)

abteilung = 'IT-Support'
print(abteilung)
```

Man nutzt oft die jeweils andere Variante, wenn der Text selbst Anführungszeichen enthält:

```
text = "Sophies Abteilung heißt 'IT-Support'."
print(text)

text = 'Er sagte: "Hallo!"'
print(text)
```

Wenn man dieselben Anführungszeichen innen und außen verwendet, gibt es einen Fehler. Lösung: andere Anführungszeichen nehmen oder mit `\\` maskieren.

```
# Falsch:
# text = "Er sagte: "Hallo!""

# Richtig:
text = 'Er sagte: "Hallo!"'
print(text)

# Oder mit Maskierung:
text = "Er sagte: \"Hallo!\""
print(text)
```

**Merksatz:**
- `'...'` und `"..."` sind für Python gleichwertig.
- Verwende innerhalb eines Strings einfach die andere Art von Anführungszeichen oder maskiere mit `\`.
- Wichtig: mit demselben Zeichen beginnen und enden – das ist die eigentliche Falle.

# Block 2: Die vier Basis-Datentypen

- **Text (String):** Muss immer in Anführungszeichen stehen.
- **Ganzzahl (Integer):** Für zählbare Dinge.
- **Nachkommazahl (Float):** In Python wird ein Punkt statt eines Kommas verwendet.
- **Wahrheitswert (Boolean):** Nur zwei Zustände (`True` oder `False`). Achtung: Großschreibung beachten.

```
abteilung = "IT-Support"       # String
tickets_geloest = 8            # Integer
preis = 49.99                  # Float
server_aktiv = True            # Boolean

print(abteilung)
print(tickets_geloest)
print(preis)
print(server_aktiv)
```

Mit `type()` kann man herausfinden, welchen Datentyp eine Variable hat:

```
a = 10
b = "Hallo"
c = 3.14
d = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

**Wichtiger Unterschied:** `5` und `"5"` sehen gleich aus, sind aber unterschiedliche Datentypen.

```
zahl1 = 5
zahl2 = "5"

print(type(zahl1))  # <class 'int'>
print(type(zahl2))  # <class 'str'>
```

# Block 3: Rechnen mit Variablen in Python

## Variablen können Ergebnisse speichern

Eine Variable kann nicht nur einen festen Wert enthalten, sondern auch das Ergebnis einer Rechnung.

> Merke: Die rechte Seite (`grundpreis + 20`) wird zuerst berechnet. Das Ergebnis wandert danach in die Variable `neuer_preis`.

```
grundpreis = 100
neuer_preis = grundpreis + 20

print(neuer_preis)
```

## Die vier Grundrechenarten

| Operator | Bedeutung | Beispiel | Ergebnis |
|:---:|:---|:---:|:---:|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraktion | `5 - 3` | `2` |
| `*` | Multiplikation | `5 * 3` | `15` |
| `/` | Division | `5 / 3` | `1.666...` |

Python nutzt `*` statt "×" und `/` statt "÷" – beides ist auf jeder Tastatur vorhanden.

```
print(5 + 3)
print(5 - 3)
print(5 * 3)
print(5 / 3)
```

## Formeln direkt übersetzen

**Schritt 1:** Formel in Worten aufschreiben → `Fläche = Breite × Länge`

**Schritt 2:** Wörter durch Variablennamen ersetzen → `flaeche = breite * laenge`

```
breite = 5
laenge = 10

flaeche = breite * laenge

print(flaeche)
```

Weiteres Beispiel: Volumen = Breite × Länge × Höhe

```
breite = 5
laenge = 10
hoehe = 2

volumen = breite * laenge * hoehe

print(volumen)
```

## Umrechnungen: kleine Einheit in große Einheit

Faustregel: Wie viele kleine Einheiten passen in eine große? → Diese Zahl ist der Multiplikator.

- 1 Stunde = 60 Minuten → `minuten = stunden * 60`
- 1 Minute = 60 Sekunden → `sekunden = minuten * 60`

```
stunden = 2
minuten = stunden * 60

print(minuten)
```

## Prozentrechnung

Prozent wird als Dezimalzahl geschrieben:

| Prozent | Als Dezimalzahl |
|:---:|:---:|
| 10 % | `0.10` |
| 15 % | `0.15` |
| 19 % | `0.19` |

**Zwei Schritte zum Endpreis:**
1. Rabattbetrag berechnen
2. Rabatt vom Preis abziehen

```
preis = 200

# Schritt 1: Rabattbetrag berechnen
rabattbetrag = preis * 0.15

# Schritt 2: Rabatt vom Preis abziehen
endpreis = preis - rabattbetrag

print(rabattbetrag)
print(endpreis)
```

## Zwischenergebnisse in eigenen Variablen speichern

Vorteil: Jeder Rechenschritt bekommt einen eigenen Namen → der Code ist leichter zu lesen, und jeder Schritt kann einzeln mit `print()` kontrolliert werden (siehe Beispiel oben).

## Von der Formel zum Code: Schritt-für-Schritt-Übersetzung

Egal wie kompliziert die Aufgabe klingt – dieser Ablauf funktioniert immer:

1. Text lesen und Zahlen markieren
2. Passende Formel finden
3. Jeder Zahl einen Variablennamen geben
4. Tabelle 1:1 in Code übertragen

| Schlüsselwort im Text | Rechenart | Operator |
|:---|:---|:---:|
| "Summe", "zusammen", "insgesamt" | Addition | `+` |
| "Differenz", "weniger", "Rabatt" | Subtraktion | `-` |
| "Fläche", "Volumen", "mal so viel" | Multiplikation | `*` |
| "aufteilen", "pro Person", "Durchschnitt" | Division | `/` |

**Beispiel:** *"Ein Raum hat eine Breite von 5 Metern und eine Länge von 8 Metern. Berechnen Sie die Fläche."*

| Text | Variablenname | Wert |
|:---|:---|:---:|
| Breite | `breite` | `5` |
| Länge | `laenge` | `8` |
| Fläche (Ergebnis) | `flaeche` | *wird berechnet* |

Tipp: Nimm immer das deutsche Wort aus der Aufgabe als Variablenname (ohne Umlaute/Leerzeichen).

```
breite = 5
laenge = 8
flaeche = breite * laenge
print(flaeche)
```

## Komplettes Beispiel: Drei Schritte nacheinander

**Aufgabe:** *"Ein Produkt kostet 200 Euro. Es gibt 15 % Rabatt. Wie hoch ist der Endpreis?"*

1. Zahlen markieren: 200, 15%
2. Formel: Rabattbetrag = Preis × Prozent, Endpreis = Preis − Rabattbetrag
3. Variablen benennen: `preis`, `rabattsatz`, `rabattbetrag`, `endpreis`

```
preis = 200
rabattsatz = 0.15

rabattbetrag = preis * rabattsatz
endpreis = preis - rabattbetrag

print(rabattbetrag)
print(endpreis)
```

- Ist die Reihenfolge korrekt (z. B. erst Rabattbetrag berechnen, dann vom Preis abziehen)?

# Block 4: String-Verknüpfung & f-Strings

## Drei Wege, Text und Variablen zu kombinieren

- **Komma `,`** → einfachste Methode, automatisches Leerzeichen
- **Plus `+`** → "Zusammenkleben", Leerzeichen selbst setzen
- **f-String `f""`** → modernste Methode, Variablen direkt in `{}`

### Methode 1: Das Komma `,`

Python setzt automatisch ein Leerzeichen zwischen den Teilen, die durch Komma getrennt sind. Das funktioniert auch problemlos mit Zahlen.

```
name = "Michael"
print("Guten Tag,", name)
```

### Methode 2: Der Plus-Operator `+`

⚠️ Beim `+` fügt Python **kein** Leerzeichen automatisch ein! Man muss es selbst als `" "` einbauen.

```
vorname = "Lisa"
nachname = "Schmid"

vollstaendiger_name = vorname + nachname
print(vollstaendiger_name)   # LisaSchmid – ohne Leerzeichen

vollstaendiger_name = vorname + " " + nachname
print(vollstaendiger_name)   # Lisa Schmid – mit Leerzeichen
```

### Die große Falle: Text + Zahl

`+` kann nur gleiche Datentypen verbinden. Text (`str`) und Zahl (`int`) lassen sich nicht einfach zusammenkleben – Python weiß nicht, ob `+` hier "addieren" oder "verketten" bedeuten soll.

| Ausdruck | Funktioniert? | Grund |
|:---|:---:|:---|
| `"Text" + "Text"` | ✅ | beides Text |
| `5 + 3` | ✅ | beides Zahl |
| `"Text" + 5` | ❌ | Text + Zahl gemischt |

```
alter = 30

# Das hier würde einen TypeError auslösen:
# print("Ich bin " + alter)

# So funktioniert es (Zahl in Text umwandeln):
print("Ich bin " + str(alter))
```

### Methode 3: f-Strings – die moderne Lösung

f-Strings verbinden Text und Zahlen ohne Fehler – der `TypeError` von Methode 2 tritt hier nie auf.

```
benutzer = "Felix"
anzahl_tickets = 4

print(f"Hallo {benutzer}, du hast heute {anzahl_tickets} Tickets gelöst.")
```

So baust du einen f-String:
1. `f` vor die Anführungszeichen setzen → `f"..."`
2. Variablen in geschweifte Klammern packen → `{variable}`
3. Alles andere bleibt normaler Text

### Vergleich aller drei Methoden

| Methode | Beispiel | Leerzeichen? | Text + Zahl mischen? |
|:---|:---|:---:|:---:|
| Komma `,` | `print("Alter:", alter)` | automatisch ✅ | ✅ problemlos |
| Plus `+` | `print("Alter: " + str(alter))` | selbst setzen ⚠️ | ❌ nur mit Umweg |
| f-String `f""` | `print(f"Alter: {alter}")` | wie normaler Text | ✅ problemlos |

🏆 Empfehlung: Wenn du Text und Zahlen mischen willst, ist der f-String fast immer die beste Wahl.

```
alter = 30
print("Alter:", alter)
print("Alter: " + str(alter))
print(f"Alter: {alter}")
```
