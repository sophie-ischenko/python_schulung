# BLOCK 3: Rechnen mit Variablen in Python



## 1️⃣ Variablen können Ergebnisse speichern

Eine Variable kann nicht nur einen festen Wert enthalten, sondern auch das **Ergebnis einer Rechnung**:


````python
grundpreis = 100

neuer_preis = grundpreis + 20

print(neuer_preis)
````


> 💡 **Merke:** Die rechte Seite (`grundpreis + 20`) wird **zuerst berechnet**. Das Ergebnis wandert danach in die Variable `neuer_preis`.

```
grundpreis + 20  ──►  120  ──►  neuer_preis
   (Rechnung)         (Ergebnis)    (Box)
```

---

## 2️⃣ Die vier Grundrechenarten

| Operator | Bedeutung | Beispiel | Ergebnis |
|:---:|:---|:---:|:---:|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraktion | `5 - 3` | `2` |
| `*` | Multiplikation | `5 * 3` | `15` |
| `/` | Division | `5 / 3` | `1.666...` |

> ⌨️ Python nutzt `*` statt "×" und `/` statt "÷" – beides ist auf jeder Tastatur vorhanden.

## 5️⃣ Prozentrechnung

### 🔑 Prozent wird als Dezimalzahl geschrieben

| Prozent | Als Dezimalzahl |
|:---:|:---:|
| 10 % | `0.10` |
| 15 % | `0.15` |
| 19 % | `0.19` |

### 🔢 Zwei Schritte zum Endpreis



````python
preis = 20

# Schritt 1: Rabattbetrag berechnen
rabattbetrag = preis * 0.15

# Schritt 2: Rabatt vom Preis abziehen
endpreis = preis - rabattbetrag
````



```
   preis          rabattbetrag         endpreis
  ┌───────┐        ┌─────────┐        ┌─────────┐
  │  200  │  × 0.15│    30   │ 200 −  │   170   │
  └───────┘  ─────►└─────────┘  ─────►└─────────┘
```

## 6️⃣ Zwischenergebnisse in eigenen Variablen speichern

> ✅ **Vorteil:** Jeder Rechenschritt bekommt einen eigenen Namen → der Code ist leichter zu lesen und jeder Schritt kann einzeln mit `print()` kontrolliert werden.



````python
preis = 200

rabattbetrag = preis * 0.15      # Schritt 1 – prüfbar
endpreis = preis - rabattbetrag  # Schritt 2 – prüfbar

print(rabattbetrag)
print(endpreis)
````

# Block 4: String-Verknüpfung & f-Strings

---

## 1️⃣ Drei Wege, Text und Variablen zu kombinieren

```
Komma  ,        →   einfachste Methode, automatisches Leerzeichen
Plus   +        →   "Zusammenkleben", Leerzeichen selbst setzen
f-String  f""   →   modernste Methode, Variablen direkt in {}
```

---

## 2️⃣ Methode 1: Das Komma `,`



````python
name = "Michael"

print("Guten Tag,", name)
````



> 💡 Python setzt **automatisch ein Leerzeichen** zwischen den Teilen, die durch Komma getrennt sind. Das funktioniert auch problemlos mit Zahlen.

```
"Guten Tag,"   ,   name
     └──────────┬──────────┘
                ▼
        Guten Tag, Michael
        (Leerzeichen automatisch)
```

---

## 3️⃣ Methode 2: Der Plus-Operator `+`



````python
vorname = "Lisa"
nachname = "Schmid"

vollstaendiger_name = vorname + nachname

print(vollstaendiger_name)

vollstaendiger_name = vorname + " " + nachname

print(vollstaendiger_name)
````



> ⚠️ **Stolperfalle:** Beim `+` fügt Python **kein** Leerzeichen automatisch ein! Man muss es selbst als `" "` einbauen.

```
vorname + nachname        →  LisaSchmid      ❌ (kein Leerzeichen)
vorname + " " + nachname  →  Lisa Schmid     ✅ (Leerzeichen selbst gesetzt)
```

### 🚫 Die große Falle: Text + Zahl


````python
alter = 30

print("Ich bin " + alter)   # ❌ TypeError!
````




> **Warum?** `+` kann nur **gleiche Datentypen** verbinden. Text (`str`) und Zahl (`int`) lassen sich nicht einfach zusammenkleben – Python weiß nicht, ob `+` hier "addieren" oder "verketten" bedeuten soll.

| Ausdruck | Funktioniert? | Grund |
|:---|:---:|:---|
| `"Text" + "Text"` | ✅ | beides Text |
| `5 + 3` | ✅ | beides Zahl |
| `"Text" + 5` | ❌ | Text + Zahl gemischt |

**Das schauen wir uns noch an**


---


## 4️⃣ Methode 3: f-Strings – die moderne Lösung




````python
benutzer = "Felix"
anzahl_tickets = 4

print(f"Hallo {benutzer}, du hast heute {anzahl_tickets} Tickets gelöst.")
````



### So baust du einen f-String:

```
Schritt 1:  f  vor die Anführungszeichen setzen        →  f"..."
Schritt 2:  Variablen in geschweifte Klammern packen    →  {variable}
Schritt 3:  Alles andere bleibt normaler Text
```

```
f"Hallo {benutzer}, du hast heute {anzahl_tickets} Tickets gelöst."
   │      └───┬───┘                  └──────┬──────┘
   │          │                             │
   │     wird ersetzt                  wird ersetzt
   │      durch "Felix"                 durch "4"
   f = Kennzeichen für f-String
```

> ✅ **Der große Vorteil:** f-Strings verbinden Text und Zahlen **ohne Fehler** – der `TypeError` von Methode 2 tritt hier nie auf!

---

## 5️⃣ Vergleich aller drei Methoden

| Methode | Beispiel | Leerzeichen? | Text + Zahl mischen? |
|:---|:---|:---:|:---:|
| Komma `,` | `print("Alter:", alter)` | automatisch ✅ | ✅ problemlos |
| Plus `+` | `print("Alter: " + str(alter))` | selbst setzen ⚠️ | ❌ nur mit Umweg |
| f-String `f""` | `print(f"Alter: {alter}")` | wie normaler Text | ✅ problemlos |

> 🏆 **Empfehlung:** Wenn du Text und Zahlen mischen willst, ist der **f-String** fast immer die beste Wahl.
