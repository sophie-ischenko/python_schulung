## 09:45 – 10:30 Uhr | Block 2: Bedingungen & Vergleichsoperatoren (Theorie)

Wir verlassen heute endgültig das Terrain des Codes, der stur von der ersten bis zur letzten Zeile durchläuft, und betreten das Feld der **Entscheidungslogik**. Das ist konzeptionell einer der wichtigsten Momente im gesamten Kurs, denn ab jetzt können unsere Programme auf unterschiedliche Situationen unterschiedlich reagieren — genau wie ein Mensch, der im Alltag ständig Entscheidungen trifft.

### 1. Konzept: Was ist Kontrollfluss?

Bisher lief jedes unserer Skripte wie ein stures Rezept ab: Zeile 1 wird ausgeführt, dann Zeile 2, dann Zeile 3 — immer in der exakt gleichen Reihenfolge, egal was passiert. Das nennt man **sequenziellen Ablauf**. In der echten Welt trifft aber praktisch jedes IT-System ständig Entscheidungen: "Ist der Speicher voll? Dann sende eine Warnung. Ist er es nicht? Dann mache normal weiter." Genau das bilden wir mit **Bedingungen** (`if`/`elif`/`else`) ab.

**Die Analogie: Die Weggabelung**
Stellen Sie sich einen Wanderer vor, der an eine Weggabelung kommt. Ohne Entscheidungslogik würde er stur immer geradeaus laufen, egal ob dort eine Straße oder ein Abgrund ist. Mit einer `if`-Bedingung kann er sich fragen: "Ist der linke Weg frei? *Wenn ja*, gehe links. *Sonst* gehe rechts." Der `if`-Block markiert also eine Weggabelung im Programmablauf: Je nachdem, ob eine Bedingung wahr (`True`) oder falsch (`False`) ist, nimmt das Programm einen anderen "Pfad" durch den Code.

**Warum ist das der Kern der Programmierung?**
Man könnte sogar sagen: Ohne Bedingungen ist ein Programm nur eine Liste von Befehlen — mit Bedingungen wird es zu einem echten "intelligenten" System, das auf seine Umgebung reagiert. Ein Monitoring-System, das bei Serverausfall automatisch eine E-Mail verschickt; eine Login-Maske, die nur bei korrektem Passwort Zugang gewährt; eine Backup-Routine, die nur startet, wenn genug Speicherplatz frei ist — all das basiert auf `if`-Bedingungen.

### 2. Syntax: Die Bausteine eines Entscheidungsblocks

```python
speicher = 8  # GB freier Speicher

if speicher < 10:
    print("Achtung: Wenig Speicherplatz!")
else:
    print("Speicherplatz ist ausreichend.")
```

Gehen wir diese wenigen Zeilen ganz genau durch, denn jedes einzelne Zeichen hat hier eine Bedeutung:

1. **Das Schlüsselwort `if`:** Es leitet die Bedingung ein und bedeutet wörtlich "wenn".
2. **Die Bedingung selbst (`speicher < 10`):** Dies ist ein sogenannter **boolescher Ausdruck** — ein Ausdruck, der von Python immer zu genau einem von zwei möglichen Werten ausgewertet wird: `True` (wahr) oder `False` (falsch). Python prüft: Ist der aktuelle Wert von `speicher` tatsächlich kleiner als 10? In unserem Beispiel ist `speicher = 8`, also lautet die Antwort `True`.
3. **Der Doppelpunkt `:`** am Ende der `if`-Zeile ist zwingend erforderlich. Er signalisiert Python: "Jetzt folgt ein Codeblock, der nur unter dieser Bedingung ausgeführt wird."
4. **Die Einrückung (Indentation):** Dies ist der Punkt, an dem die meisten Python-Einsteiger stolpern, wenn sie von anderen Sprachen kommen. Anders als z. B. in Java oder C++, wo geschweifte Klammern `{}` einen Codeblock markieren, verwendet Python **die Einrückung selbst** als Syntaxelement. Alle Zeilen, die zu einem `if`-Block gehören, müssen um exakt die gleiche Anzahl an Leerzeichen eingerückt sein (üblicherweise 4 Leerzeichen, in Colab automatisch durch Drücken von Tab). Fehlt diese Einrückung oder ist sie inkonsistent (mal 2, mal 4 Leerzeichen), wirft Python einen `IndentationError`.
5. **Das Schlüsselwort `else`:** Es bedeutet "andernfalls" und fängt exakt den Fall ab, in dem die `if`-Bedingung `False` ergeben hat. Ein `else`-Block ist optional — man kann auch ein `if` ganz ohne `else` schreiben, dann passiert im "falschen" Fall einfach gar nichts.

**Wichtig zu verstehen:** Python prüft die Bedingung **genau einmal**, an genau der Stelle, an der der Code ausgeführt wird. Wenn sich `speicher` später im Programm ändert, wird die `if`-Bedingung nicht automatisch neu geprüft — man müsste sie an der entsprechenden Stelle im Code erneut aufrufen.

### 3. Vergleichsoperatoren — das Vokabular der Bedingungen

Um überhaupt eine Bedingung formulieren zu können, brauchen wir eine Möglichkeit, zwei Werte miteinander zu vergleichen. Dafür stellt Python folgende **Vergleichsoperatoren** bereit, die jeweils zu `True` oder `False` ausgewertet werden:

| Operator | Bedeutung | Beispiel | Ergebnis |
|---|---|---|---|
| `==` | Gleichheit | `5 == 5` | `True` |
| `!=` | Ungleichheit | `5 != 3` | `True` |
| `<` | Kleiner als | `3 < 5` | `True` |
| `>` | Größer als | `3 > 5` | `False` |
| `<=` | Kleiner oder gleich | `5 <= 5` | `True` |
| `>=` | Größer oder gleich | `4 >= 5` | `False` |

**Die wichtigste Stolperfalle überhaupt:** Der Unterschied zwischen `=` und `==`. Ein einzelnes Gleichheitszeichen `=` ist in Python die **Zuweisung** — es packt einen Wert in eine Variable (`x = 5` bedeutet "speichere die Zahl 5 in der Variable x"). Ein doppeltes Gleichheitszeichen `==` ist der **Vergleichsoperator** — er fragt "sind diese beiden Werte gleich?" und liefert `True` oder `False` zurück, verändert aber keine Variable. Verwechselt man beide in einer `if`-Bedingung, führt das entweder zu einem `SyntaxError` oder — schlimmer noch — zu einem stillen Logikfehler, den man erst sehr viel später bemerkt.

````python
# Mitmach-Beispiel: if / else mit Vergleichsoperatoren
speicher = 8  # GB freier Speicher

if speicher < 10:
    print("Achtung: Wenig Speicherplatz!")
else:
    print("Speicherplatz ist ausreichend.")
````

## 10:45 – 11:30 Uhr | Block 3: Logische Operatoren & Listen-Mitgliedschaften (Theorie)

Eine einzelne Bedingung reicht in der Praxis oft nicht aus. Ein Server ist nicht nur dann kritisch überlastet, wenn die CPU-Auslastung hoch ist — sondern womöglich erst dann, wenn *gleichzeitig* auch der Arbeitsspeicher knapp wird. Für solche Situationen brauchen wir eine Möglichkeit, mehrere Bedingungen **logisch miteinander zu verknüpfen**.

### 1. Logische Operatoren — Bedingungen kombinieren

Python stellt dafür drei sogenannte **logische Operatoren** (auch "boolesche Operatoren" genannt) bereit:

**`and` (Und-Verknüpfung):** Das Gesamtergebnis ist nur dann `True`, wenn **beide** verknüpften Bedingungen für sich genommen `True` sind. Ist auch nur eine der beiden Bedingungen `False`, ist das Gesamtergebnis ebenfalls `False`. Stellen Sie sich das wie eine Tür mit zwei Schlössern vor, die beide gleichzeitig aufgeschlossen sein müssen, damit die Tür aufgeht.

**`or` (Oder-Verknüpfung):** Das Gesamtergebnis ist bereits dann `True`, wenn **mindestens eine** der beiden Bedingungen `True` ist. Nur wenn *beide* Bedingungen `False` sind, ist auch das Gesamtergebnis `False`. Das entspricht eher einer Tür mit zwei unabhängigen Klingelknöpfen: Es reicht, wenn *irgendjemand* klingelt, damit die Glocke ertönt.

**`not` (Verneinung):** Dieser Operator dreht einen einzelnen Wahrheitswert einfach um: Aus `True` wird `False`, und aus `False` wird `True`. Er wird oft verwendet, um eine Bedingung "umzudrehen", ohne den eigentlichen Vergleich neu schreiben zu müssen — z. B. `not gesperrt` bedeutet "das System ist *nicht* gesperrt".

Eine hilfreiche Merktabelle (die sogenannte "Wahrheitstabelle") für `and` und `or`:

| Bedingung A | Bedingung B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

In der Praxis der IT-Administration begegnen einem solche Verknüpfungen ständig: "Sende nur dann einen Alarm, wenn die CPU-Auslastung über 80% **und** die RAM-Auslastung über 80% liegt" (beides muss gleichzeitig zutreffen, sonst wäre ein einzelner Ausreißer schon ein Fehlalarm). Oder: "Zeige eine Warnung, wenn die CPU-Auslastung hoch **oder** die RAM-Auslastung hoch ist" (hier reicht bereits eine einzelne Komponente, um vorsichtig zu werden).

**Gemeinsames Live-Coding:**

````python
# Live-Coding: Logische Operatoren and / or
cpu_auslastung = 85
ram_auslastung = 90

# Beide müssen kritisch sein
if cpu_auslastung > 80 and ram_auslastung > 80:
    print("Kritisch: Gesamter Server ist überlastet!")

# Einer von beiden reicht
if cpu_auslastung > 80 or ram_auslastung > 80:
    print("Warnung: Eine Komponente ist stark ausgelastet.")
````

### 2. Mitgliedschaften prüfen mit `in`

Ein weiteres extrem nützliches Werkzeug, das perfekt mit `if`-Bedingungen zusammenspielt, ist der `in`-Operator, den Sie an Tag 4 bereits im Zusammenhang mit Listen kennengelernt haben. Er prüft, ob ein bestimmter Wert **Mitglied** einer Liste (oder eines Strings) ist, und liefert ebenfalls direkt `True` oder `False` zurück. Kombiniert mit einer `if`-Bedingung lässt sich damit sehr elegant prüfen, ob ein bestimmter Fall in einer vordefinierten Liste "erlaubter" oder "gesperrter" Werte vorkommt — zum Beispiel bei einer VIP-Kundenliste, einer Blacklist von IP-Adressen oder einer Liste zulässiger Benutzernamen.

Der große Vorteil gegenüber dem manuellen Vergleich jedes einzelnen Elements (`kunde == "Müller" or kunde == "Schmidt" or kunde == "Fischer"`) ist die Übersichtlichkeit: Eine einzige `in`-Prüfung gegen eine Liste ersetzt potenziell dutzende einzelne `or`-Verknüpfungen und lässt sich viel leichter erweitern — man muss der Liste einfach nur einen weiteren Namen hinzufügen.

```python
vip_kunden = ["Müller", "Schmidt", "Fischer"]
kunde = "Schmidt"

if kunde in vip_kunden:
    print("Priorisierter VIP-Support aktivieren.")
else:
    print("Standard-Support-Warteschlange.")
```

````python
# Live-Coding: Mitgliedschaften prüfen mit in
vip_kunden = ["Müller", "Schmidt", "Fischer"]
kunde = "Schmidt"

if kunde in vip_kunden:
    print("Priorisierter VIP-Support aktivieren.")
else:
    print("Standard-Support-Warteschlange.")
````

Bevor wir weitermachen, kurz zum neuen Baustein `elif` (kurz für "else if"): Er erlaubt es, **mehr als zwei** mögliche Fälle nacheinander zu prüfen.
