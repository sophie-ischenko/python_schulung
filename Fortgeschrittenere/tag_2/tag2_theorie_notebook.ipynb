## 2.1 Funktionen & Parameter

**Definition:** Eine Funktion ist ein benannter, wiederverwendbarer Codeblock, der eine klar abgegrenzte Aufgabe erfuellt. Parameter sind Platzhalter fuer Werte, die beim Aufruf uebergeben werden.

**Warum wichtig:** Ohne Funktionen wiederholt sich Logik (z.B. Schaden wuerfeln) an mehreren Stellen im Code - jede Aenderung muesste dann mehrfach gepflegt werden.

**Syntax:**
```python
def funktionsname(parameter1, parameter2):
    """Kurzbeschreibung: Was / Parameter / Rueckgabewert"""
    anweisungen
```

**Merksatz (gilt fuer den Rest des Kurses):** *"Jede Funktion beantwortet drei Fragen: Was macht sie? Was nimmt sie entgegen? Was gibt sie zurueck?"*

**Haeufiger Fehler:** Eine Funktion ohne `return` liefert automatisch `None` zurueck - ein klassischer Stolperstein ist, das Ergebnis eines `print()`-Aufrufs versehentlich weiterverwenden zu wollen, obwohl `print()` selbst nichts zurueckgibt.

---
## 2.2 Rueckgabewerte (return)

**Definition:** `return` beendet eine Funktion sofort und uebergibt einen Wert an die aufrufende Stelle, statt ihn nur anzuzeigen.

**Warum wichtig:** `print()` gibt einen Wert nur auf dem Bildschirm aus - danach ist er fuer das Programm "weg". `return` macht das Ergebnis weiterverarbeitbar, z.B. um es spaeter statt auf der Konsole in einem GUI-Label anzuzeigen (Tag 5).

**Syntax:**
```python
def funktionsname(parameter):
    ...
    return ergebnis
```

**Wichtige Nuance:** Sobald Python auf ein `return` trifft, wird die Funktion sofort verlassen - jeglicher Code danach in derselben Funktion wird nicht mehr ausgefuehrt. Das laesst sich gezielt nutzen, um fruehzeitig aus einer Funktion auszusteigen ("early return"), sobald ein Ergebnis feststeht.

**Merksatz:** *"print() zeigt, return liefert. Logik-Funktionen sollten fast immer return nutzen, Anzeige-Funktionen print."*

## 2.3 Dateien & with

**Definition:** `open()` oeffnet eine Datei zum Lesen (`"r"`) oder Schreiben (`"w"`). Der `with`-Block sorgt dafuer, dass die Datei automatisch geschlossen wird - auch wenn ein Fehler auftritt.

**Warum wichtig:** Ohne dauerhafte Speicherung geht jeder Spielstand beim Beenden des Programms verloren.

**Syntax:**
```python
with open("pfad", "modus", encoding="utf-8") as datei:
    ...
```

**Die wichtigsten Modi:**

| Modus | Bedeutung |
|---|---|
| `"r"` | Lesen (Datei muss existieren) |
| `"w"` | Schreiben (ueberschreibt vorhandenen Inhalt komplett!) |
| `"a"` | Anhaengen (haengt ans Dateiende an, statt zu ueberschreiben) |

**Merksatz:** *"Immer `with` statt manuellem open()/close() - sonst riskiert man vergessene, gesperrte Dateien."*

---
## 2.4 JSON

**Definition:** JSON wandelt ein Python-Dictionary in speicherbaren Text um (`json.dump`) und wieder zurueck (`json.load`).

**Warum wichtig:** Ein Charakter besteht aus mehreren zusammengehoerigen Werten (Name, HP, Level) - JSON speichert dieses ganze Paket in einer Datei, nicht nur eine einzelne Zahl.

**Syntax:**
```python
json.dump(daten, datei)      # Dictionary -> Datei
geladen = json.load(datei)   # Datei -> Dictionary
```

**Haeufiger Fehler:** `json.load()` scheitert mit einem `FileNotFoundError`, wenn die Datei noch nicht existiert (z.B. beim allerersten Programmstart). Deshalb wird das Laden ueblicherweise in `try/except FileNotFoundError` eingebettet, mit einem sinnvollen Standardwert im `except`-Zweig.

**Merksatz:** *"CSV/reiner Text fuer einzelne Werte, JSON fuer zusammengehoerige Datenpakete."*
