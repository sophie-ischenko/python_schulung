# Block 1: Funktionsaufrufe festigen (09:00 – 09:45 Uhr)

## Theorie: Der Wertfluss zwischen Hauptprogramm und Funktion

Du hast an Tag 8 gelernt, was `return` von `print` unterscheidet. Heute schaust du dir noch genauer an, **wie** ein Wert überhaupt von deinem Hauptprogramm in eine Funktion hinein- und wieder herausfließt. Dieses Verständnis ist die Grundlage für alles, was heute noch kommt – vor allem für Block 2, in dem Funktionen sich gegenseitig aufrufen.

### Visualisierung des Wertflusses

Wenn eine Funktion aufgerufen wird, springt Python an die Stelle im Code, an der die Funktion definiert wurde, arbeitet diese ab und bringt das Ergebnis (`return`) zurück an die Stelle des Aufrufs.

```
Hauptprogramm:  ergebnis = berechne_wert(10)   <-- 1. Übergibt 10
                                    |
                                    v
Definition:     def berechne_wert(zahl):       <-- 2. Empfängt 10 in 'zahl'
                    return zahl * 2            <-- 3. Berechnet 20 und liefert zurück
                                    |
                                    v
Hauptprogramm:  ergebnis = 20                  <-- 4. Speichert 20 in 'ergebnis'
```

Wichtig dabei: Der Parametername (`zahl`) ist nur ein **Platzhalter innerhalb der Funktion**. Er hat mit dem Namen der Variable, die du beim Aufruf übergibst, nichts zu tun. Du könntest genauso gut schreiben:

```python
mein_wert = 10
ergebnis = berechne_wert(mein_wert)
```

Python kümmert sich automatisch darum, den *Wert* von `mein_wert` (also `10`) in den Parameter `zahl` zu kopieren. Innerhalb der Funktion arbeitest du dann mit dem Namen `zahl`, ganz unabhängig davon, wie die Variable draußen im Hauptprogramm hieß.

### Warum das wichtig ist

Genau dieses Prinzip macht Funktionen so mächtig: Du kannst dieselbe Funktion mit ganz unterschiedlichen Variablen (oder auch direkt mit festen Werten) aufrufen, ohne dass sich die Funktion selbst je ändern muss. Sie "kennt" nur ihre Parameter – nicht, woher die Werte stammen.

# Block 2: Funktionen rufen Funktionen (09:45 – 10:30 Uhr)

## Theorie: Modularisierung

In der professionellen Softwareentwicklung schreibt man keine riesigen "Monolithen" – also riesige Funktionen, die alles auf einmal tun (Eingabe prüfen, berechnen, formatieren, ausgeben). Das ist unübersichtlich und schwer zu warten: Wenn irgendwo ein Fehler steckt, musst du dich durch hunderte Zeilen wühlen.

Stattdessen zerlegt man komplexe Probleme in kleine, **hochspezialisierte Funktionen**, die jeweils nur eine einzige, klar abgegrenzte Aufgabe erledigen. Der Clou dabei: Eine Funktion kann eine **andere Funktion aufrufen**, um einen Teilschritt zu erledigen. So baust du dir Stück für Stück größere Programme aus kleinen, gut testbaren Bausteinen zusammen. Man nennt dieses Vorgehen **Modularisierung**.

### Warum das wichtig ist

- **Testbarkeit:** Du kannst jede kleine Funktion einzeln testen, statt ein riesiges Konstrukt auf einmal prüfen zu müssen.
- **Lesbarkeit:** Der Name einer Funktion (z. B. `ermittle_steuer`) verrät sofort, was sie tut – du musst nicht erst den ganzen Code lesen, um zu verstehen, was passiert.
- **Wiederverwendbarkeit:** Eine kleine Hilfsfunktion wie `ermittle_steuer` kannst du später in einem ganz anderen Zusammenhang wiederverwenden.

## Live-Coding: Der Bestell-Prozess für Hardware

Schau dir an, wie zwei spezialisierte Hilfsfunktionen von einer dritten, übergeordneten Funktion aufgerufen werden, um ein Gesamtergebnis zu berechnen.


````
# Hilfsfunktion 1: Berechnet die Mehrwertsteuer
def ermittle_steuer(netto_betrag):
    return netto_betrag * 0.19

# Hilfsfunktion 2: Berechnet den Rabatt
def ermittle_rabatt(netto_betrag):
    if netto_betrag > 1000:
        return netto_betrag * 0.10  # 10% Rabatt bei Großbestellung
    return 0.0

# Haupt-Kalkulationsfunktion (ruft die Hilfsfunktionen auf!)
def kalkuliere_gesamtpreis(netto_betrag):
    steuer = ermittle_steuer(netto_betrag)
    rabatt = ermittle_rabatt(netto_betrag)

    brutto_gesamt = netto_betrag + steuer - rabatt
    return brutto_gesamt

# Aufruf
endpreis = kalkuliere_gesamtpreis(1200)
print(f"Endpreis nach Rabatt und Steuer: {endpreis} Euro")
````

**Was passiert hier genau?**

1. `kalkuliere_gesamtpreis` wird mit `1200` aufgerufen.
2. Innerhalb dieser Funktion wird zuerst `ermittle_steuer(netto_betrag)` aufgerufen – das ist ein ganz normaler Funktionsaufruf, nur dass er *innerhalb* einer anderen Funktion steht, statt im Hauptprogramm.
3. Das Ergebnis dieses Aufrufs wird in `steuer` gespeichert.
4. Genauso wird `ermittle_rabatt(netto_betrag)` aufgerufen und das Ergebnis in `rabatt` gespeichert.
5. Erst danach werden beide Zwischenergebnisse zum Endpreis verrechnet und zurückgegeben.

Für `kalkuliere_gesamtpreis` spielt es dabei überhaupt keine Rolle, *wie* `ermittle_steuer` oder `ermittle_rabatt` ihre Berechnung im Detail durchführen – sie "vertraut" einfach darauf, dass ein sinnvolles Ergebnis zurückkommt. Das nennt man **Abstraktion**: Details werden in Hilfsfunktionen versteckt, damit sich die übergeordnete Funktion auf das große Ganze konzentrieren kann.

### Merksatz

> Eine Funktion, die andere Funktionen aufruft, ist wie ein Vorarbeiter, der Teilaufgaben an Spezialisten delegiert, statt alles selbst zu erledigen.

### Kurze Experimentier-Zelle

````
# Probiere es selbst aus:
print(kalkuliere_gesamtpreis(500))
print(kalkuliere_gesamtpreis(1500))
````

# Block 3: Die `main()`-Funktion und der Einstiegspunkt (10:45 – 11:30 Uhr)

## Theorie: Warum `def main():`?

Bisher hast du Variablen meist direkt "global" im Notebook definiert – also außerhalb jeder Funktion, sozusagen frei im gesamten Dokument sichtbar. In kleinen Übungen ist das kein Problem. In echten, größeren Python-Skripten führt das aber schnell zu Unordnung: Man spricht davon, dass der **globale Namensraum "verschmutzt"** wird – es wird immer unübersichtlicher, welche Variable wo herkommt und von wem sie verändert werden kann.

**Der Standard in der professionellen Python-Entwicklung:** Du packst den gesamten Steuerungs-Ablauf deines Programms – also die Benutzerinteraktion und die Aufrufe der anderen Funktionen – in eine eigene Funktion namens `main()` (englisch für "Haupt-"). Damit hat dein Programm einen einzigen, klar erkennbaren Ausgangspunkt.

## Theorie: Das Python-Einstiegspunkt-Pattern

Am Ende eines Python-Skripts findet man deshalb fast immer folgende Konstruktion:

```python
def main():
    # Hier startet das eigentliche Programm
    print("Das Hauptprogramm läuft...")

# Der Einstiegspunkt
if __name__ == "__main__":
    main()
```

### Was bedeutet `if __name__ == "__main__":`?

Jede Python-Datei besitzt intern eine automatisch gesetzte Variable namens `__name__`. Wird die Datei **direkt ausgeführt** (also z. B. mit `python mein_skript.py` gestartet), hat `__name__` automatisch den Wert `"__main__"`. Wird dieselbe Datei jedoch von einem **anderen** Skript nur als Baustein importiert (`import mein_skript`), hat `__name__` einen anderen Wert – nämlich den Dateinamen.

Die Zeile `if __name__ == "__main__":` sorgt also dafür, dass die Funktion `main()` **nur dann automatisch ausgeführt wird, wenn diese Datei direkt gestartet wird** – nicht aber, wenn sich jemand nur die darin definierten Funktionen "ausleihen" möchte, ohne dass das ganze Programm gleich losläuft.

Das ist der eigentliche Übergang vom reinen "Notebook-Basteln", bei dem Code einfach von oben nach unten abgearbeitet wird, hin zur professionellen Softwareentwicklung, bei der jede Datei sowohl eigenständig lauffähig als auch als Baukasten für andere Dateien nutzbar ist.

### Merksatz

> `main()` bündelt den Ablauf deines Programms an einer Stelle. `if __name__ == "__main__":` sorgt dafür, dass dieser Ablauf nur startet, wenn du die Datei auch wirklich selbst ausführst.

### Hinweis für Colab

In Google Colab kannst du das Pattern genauso schreiben und verwenden wie in einem lokalen Skript – der Aufruf `main()` am Ende sorgt dafür, dass dein Programm tatsächlich losläuft, sobald du die Zelle ausführst.


````
def main():
    print("Das Hauptprogramm läuft...")

if __name__ == "__main__":
    main()
````

*   **Schritt 2:** Schreibe eine Funktion `erzeuge_alarm_text(status, servername)`. Wenn der Status `"kritisch"` ist, soll sie den Text `"[WARNUNG] Server [Name] ist überhitzt!"` zurückgeben. Andernfalls: `"[INFO] Server [Name] läuft stabil."`.

````
# Lösung Programm-System 1
def bewerte_temperatur(temp):
    if temp > 30:
        return "kritisch"
    return "normal"

def erzeuge_alarm_text(status, servername):
    if status == "kritisch":
        return f"[WARNUNG] Server {servername} ist überhitzt!"
    return f"[INFO] Server {servername} läuft stabil."

def main():
    name = input("Servername: ")
    temp = float(input("Temperatur in °C: "))

    status = bewerte_temperatur(temp)
    meldung = erzeuge_alarm_text(status, name)
    print(meldung)

if __name__ == "__main__":
    main()
````

### Programm-System 3: Der Ping-Latenz-Tester (mit Schleife)

Jetzt kombinierst du Funktionen, die `main()`-Struktur und `while`-Schleifen aus Tag 8 miteinander.

*   **Schritt 1:** Schreibe eine Funktion `bewerte_latenz(ping)`.
    *   Ping <= 30 ms: Rückgabe `"Hervorragend"`
    *   Ping <= 100 ms: Rückgabe `"Akzeptabel"`
    *   Darüber: Rückgabe `"Kritisch"`
*   **Schritt 2:** Schreibe eine `main()`-Funktion. Diese soll dich in einer `while`-Schleife immer wieder nach einer Ping-Latenz (Ganzzahl) fragen.
    *   Gibst du `-1` ein, bricht die Schleife (`break`) ab.
    *   Bei allen anderen Eingaben wird die Latenz über `bewerte_latenz` bewertet und das Ergebnis direkt ausgegeben.


````
# Deine Lösung für Programm-System 3:

````

<details>
<summary>💡 Lösung Programm-System 3 anzeigen</summary>

Schau erst rein, wenn du es selbst versucht hast!
</details>


````
# Lösung Programm-System 3
def bewerte_latenz(ping):
    if ping <= 30:
        return "Hervorragend"
    elif ping <= 100:
        return "Akzeptabel"
    else:
        return "Kritisch"

def main():
    print("--- Latenz-Überwachungstool (Eingabe -1 zum Beenden) ---")
    while True:
        eingabe = int(input("Gemessene Latenz in ms: "))
        if eingabe == -1:
            print("Tool beendet.")
            break

        status = bewerte_latenz(eingabe)
        print(f"Latenz-Bewertung: {status}")

if __name__ == "__main__":
    main()
````

Besprich mit der Gruppe: Warum ist der heutige modulare Code einfacher zu warten als ein einzelner, langer Codeblock, der alles auf einmal macht? Ein paar Gedanken dazu:

- **Fehlersuche wird einfacher:** Wenn z. B. die generierte E-Mail-Adresse falsch aussieht, weißt du sofort, dass der Fehler in `generiere_email` liegen muss – du musst nicht das ganze Programm durchsuchen.

- **Die `main()`-Funktion liest sich fast wie eine Inhaltsangabe:** Wer dein Programm zum ersten Mal liest, kann in `main()` auf einen Blick erkennen, *was* passiert (Eingabe → Bewertung → Ausgabe), ohne sich sofort mit *wie* jeder einzelne Schritt berechnet wird, beschäftigen zu müssen.
